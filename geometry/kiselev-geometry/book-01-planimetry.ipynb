## Introduction

Basic concepts:
- **Point** - has position but no size
- **Line** - extends infinitely in both directions
- **Segment** - portion of a line between two points
- **Ray** - line extending from a point in one direction

These are the building blocks for everything that follows.

## Chapter 1: The Straight Line

### §1 Angles

An **angle** is formed by two rays sharing a common endpoint (the vertex).

Key concepts:
- **Right angle** = 90° (quarter turn)
- **Straight angle** = 180° (half turn)
- **Supplementary angles** - two angles that sum to 180°
- **Complementary angles** - two angles that sum to 90°

---

### §2 Perpendicular Lines

Two lines are **perpendicular** if they intersect at a right angle (90°).

Notation: $AB \perp CD$ means line AB is perpendicular to line CD.

---

## Key Theorem: Uniqueness of Perpendicular from External Point

**Proposition 24:** From any point M lying outside a given line AB, one can drop a perpendicular to this line, and such perpendicular is **unique**.

### The Proof (Using Folding)

1. Given: Line AB and point M not on the line
2. Fold the diagram so the upper part meets the lower part
3. Point M moves to position N (its mirror image across AB)
4. Let C be where line MN crosses AB

**Claim:** MN ⊥ AB, and this is the only perpendicular from M.

**Why MN is perpendicular:**
- After folding, M lands on N
- C stays in place (it's on the fold line AB)
- So MC folds onto NC
- Therefore ∠MCB = ∠NCB (the fold preserves angles)
- But ∠MCB and ∠NCB are supplementary (they form a straight line MN)
- Two equal supplementary angles must each be 90°
- Therefore MN ⊥ AB ✓

**Why no other line through M is perpendicular:**

Consider any other line MD (where D ≠ C):
- After folding, M → N but D stays in place
- MD folds onto ND
- So ∠MDC = ∠NDC
- But MDN is NOT a straight line (because there's only one straight line from M to N)
- So ∠MDC + ∠NDC ≠ 180°
- Therefore ∠MDC ≠ 90°
- Therefore MD is NOT perpendicular to AB ✓

---

### My Intuition: The "Path to a Friend" Analogy

I like to think about this in terms of walking to a friend's house:

**Setup:**
- You're at point M
- Your friend is at point N (directly across the line AB)
- The line AB is a road you must cross

**The perpendicular (through C):**

If the crossing point C is the midpoint between M and N, then you walk in a **straight line** from M to N. No turning required. The angle at C is zero degrees of "turn" — which means you're crossing at 90° to the road.

**Any other path (through D):**

If you cross at some other point D (not the midpoint), your path MDN is NOT straight. You have to **turn** at D to reach N. This turning angle is greater than zero, which means you're NOT crossing perpendicular to the road.

**The key insight:**

There's only ONE straight line from M to N (basic axiom: two points determine a unique line). This line crosses AB at exactly one point C. That point C is where the perpendicular drops.

Any other crossing point D requires a "bent" path → not perpendicular.

---

### Visual Summary

```
         M
         |
         | (perpendicular - straight path)
         |
----A----C----B---- (the line)
         |
         |
         N

vs.

         M
          \
           \  (bent path - NOT perpendicular)
            \
----A----D----B----
           /
          /
         N
```

The straight path M→C→N is the perpendicular. Any bent path M→D→N is not.

## §4–§5: Congruence Tests for Triangles

Two triangles are **congruent** if one can be placed exactly on top of the other — every side and every angle matches.

### The three tests (summary)

| Test | What you know | How it is proved |
|------|---------------|-----------------|
| **SAS** (Test 1) | Two sides + the angle between them | Superimposition (slide and rotate) |
| **ASA** (Test 2) | Two angles + the side between them | Superimposition |
| **SSS** (Test 3) | All three sides equal | Juxtaposition (back-to-back) — see below |

---

### Why SSS cannot be proved by superimposition

For SAS and ASA we slid one triangle onto the other. As soon as two sides coincide and the angle between them is forced, the third side snaps into place automatically.

With SSS, we only know lengths — no angles given. If we slid triangle $ABC$ onto $A'B'C'$ so that $A'C'$ coincides with $AC$, we would still have no idea whether $B$ and $B'$ end up at the same point.

**The fix:** instead of sliding, we glue the triangles back-to-back along their equal side, then use isosceles triangles to sneak in the angle equality we need.

---

### SSS Proof — Step by step

**Given:** $\triangle ABC$ and $\triangle A'B'C'$ with $AB = A'B'$, $BC = B'C'$, $CA = C'A'$.

**Goal:** Show the triangles are congruent.

**Step 1 — Juxtapose (glue back-to-back)**

Lay $\triangle A'B'C'$ so that side $A'C'$ coincides exactly with $AC$, but with $B'$ on the *opposite* side of $AC$ from $B$. Call the relocated $B'$ the point $B''$.

```
           B
          /|\
         / | \
        /  |  \
       /   |   \
      A----+----C
       \   |   /
        \  |  /
         \ | /
          \|/
           B'' (= relocated B')
```

We now have a diamond (quadrilateral) $A B C B''$.

**Step 2 — Draw the auxiliary line $B'B''$ (which is $BB''$ in our relabelled diagram)**

This diagonal cuts the diamond into two triangles:
- Top: $\triangle B' A' B''$ (or equivalently $\triangle B A B''$)
- Bottom: $\triangle B' C' B''$ (or equivalently $\triangle B C B''$)

**Step 3 — Identify the two isosceles triangles**

- $AB = A'B' = AB''$ (given, both equal $AB$) → $\triangle BAB''$ is **isosceles** with apex $A$
- $CB = C'B' = CB''$ (given, both equal $CB$) → $\triangle BCB''$ is **isosceles** with apex $C$

**Step 4 — Equal base angles**

In any isosceles triangle, base angles are equal:
$$\angle 1 = \angle 2 \quad \text{(base angles of } \triangle BAB''\text{)}$$
$$\angle 3 = \angle 4 \quad \text{(base angles of } \triangle BCB''\text{)}$$

**Step 5 — Combine to get the full angle equality**

$$\angle ABC = \angle 1 + \angle 3 = \angle 2 + \angle 4 = \angle AB''C = \angle A'B'C'$$

So $\angle B = \angle B'$.

**Step 6 — Apply SAS**

Now we have:
- $AB = A'B'$ (given)
- $\angle B = \angle B'$ (just proved)
- $BC = B'C'$ (given)

By the SAS test, $\triangle ABC \cong \triangle A'B'C'$. $\square$

---

### My intuition for why this works

The juxtaposition trick works because two isosceles triangles sharing the same base ($BB''$) act like a "mirror check". If the sides match, the base angles must match. And if the base angles on both sides of the dividing line match, the full angles at $B$ and $B''$ must be equal.

You cannot just slide the triangles together because you don't know yet whether $B$ and $B'$ would land in the same place. By putting them on opposite sides you avoid that question — and instead let the isosceles property answer it for you.

## §7: Inequalities in Triangles

### Key definition — Exterior angle

Extend one side of the triangle past a vertex. The angle formed between the extended side and the remaining side is the **exterior angle**.

```
        A
       /|
      / |
     /  |
    /   |
   B----C---------D
        ↑
      exterior angle ∠BCD
      (supplementary to interior ∠BCA)
```

- Interior angle at $C$: $\angle BCA$
- Exterior angle: $\angle BCD$
- They form a straight line, so $\angle BCA + \angle BCD = 180°$

The interior angles **not supplementary** to $\angle BCD$ are $\angle A$ and $\angle B$ — the two angles on the far side of the triangle.

---

### Theorem 42 (Kiselev)

> **An exterior angle of a triangle is greater than each non-adjacent interior angle.**
>
> $\angle BCD > \angle A \quad$ and $\quad \angle BCD > \angle B$

---

### Kiselev's original proof (brief reference)

Kiselev finds the midpoint $E$ of $BC$, draws the median $AE$, extends it by an equal length to point $F$, then shows $\triangle ABE \cong \triangle FCE$ (by SAS at the shared vertex $E$). From congruence, $\angle B = \angle ECF$. Since $\angle ECF$ is only part of $\angle BCD$, we get $\angle B < \angle BCD$.

This proof works even without using the $180°$ angle sum — it is valid in any "neutral" geometry.

---

### My own proof — using the Angle Sum Theorem

> *(This proof uses the Euclidean fact that angles in a triangle sum to $180°$.  
> It is cleaner and more direct than Kiselev's auxiliary-point construction.)*

**Setup**

Label the angles:
$$\angle A + \angle B + \angle C = 180° \quad \text{(angle sum of triangle)}$$

The exterior angle $\angle BCD$ and the interior angle $\angle C$ lie on a straight line:
$$\angle C + \angle BCD = 180° \quad \text{(supplementary angles on line BD)}$$

**The key step**

Both expressions equal $180°$, so set them equal:
$$\angle A + \angle B + \angle C = \angle C + \angle BCD$$

Cancel $\angle C$ from both sides:
$$\boxed{\angle BCD = \angle A + \angle B}$$

This is the **Remote Exterior Angle Theorem**.

**Finish**

In any real triangle, every angle is strictly greater than $0°$.
Therefore:
$$\angle BCD = \angle A + \angle B > \angle A \quad \checkmark$$
$$\angle BCD = \angle A + \angle B > \angle B \quad \checkmark$$

The exterior angle is strictly larger than either remote interior angle because it equals *their sum*, and both summands are positive. $\square$

---

### Rotating-line visualisation

**The core idea:** fix side $AC$ in place, and imagine swinging side $BC$ around the pivot $C$ like a clock hand.

As $BC$ rotates, the exterior angle $\angle BCD$ changes. Watch what happens to $\angle A$ and $\angle B$:

```
FRAME 1 — exterior angle is small (90°)

        A
        |  B
        | /
        |/ ∠C = 90°
        C-----------D
        ←  ∠BCD = 90° →

   ∠A + ∠B = 90°   (they share the 90° pool)
   Neither ∠A nor ∠B can equal 90° — they must split it.


FRAME 2 — exterior angle is medium (120°)

        A
        |
        |   B
        |  /
        | / ∠C = 60°
        C-----------D
        ←  ∠BCD = 120° →

   ∠A + ∠B = 120°   (bigger pool, but still shared)
   Neither can be 120° — they both contribute.


FRAME 3 — exterior angle is large (160°)

        A
        |
        |
        |      B
        |     /
        |    / ∠C = 20°
        C-----------D
        ←  ∠BCD = 160° →

   ∠A + ∠B = 160°   (almost the whole straight angle)
   Neither is 160° — they share even this large pool.
```

**The pattern:**

No matter how you rotate $BC$, the exterior angle $\angle BCD$ acts as a container that holds the combined value of $\angle A$ and $\angle B$. Since both $\angle A$ and $\angle B$ are always positive, neither one can fill the whole container — each is strictly smaller than the total.

```
                        ┌─────────────────────────────────────┐
   Exterior angle  =    │    ∠A     │         ∠B              │
    ∠BCD = ∠A + ∠B      └──────────────────────────────────── ┘

   ∠A alone fills only part of the bar → ∠A < ∠BCD
   ∠B alone fills only part of the bar → ∠B < ∠BCD
```

---

### Why my proof is valid (and when Kiselev's is needed)

| | My proof | Kiselev's proof |
|---|---|---|
| Uses angle sum = 180°? | Yes | No |
| Works in Euclidean geometry? | Yes | Yes |
| Works in hyperbolic geometry? | No | Yes |
| Simpler and more direct? | Yes | No |
| Good for an intro course? | Yes | Yes (teaches neutral geometry) |

For studying Kiselev and standard Euclidean geometry, my proof is perfectly correct. Kiselev's construction-based proof is more powerful because it avoids assuming the parallel postulate, but my version makes the intuition immediate.

---

### Core takeaway

$$\text{exterior angle} = \text{sum of the two non-adjacent interior angles}$$

This one equation tells you everything:
- The exterior angle is strictly greater than each remote angle (because it is their sum and both are positive).
- The exterior angle grows as the interior angle $C$ shrinks (they are supplementary).
- The remote angles $A$ and $B$ always share the exterior angle — neither can dominate alone.
