**Book:** Gelfand & Saul — *Trigonometry*

## Problems 2 & 3 — From an altitude to the Law of Sines

An acute triangle $ABC$ with the altitude $h$ from $C$ meeting $AB$ at $D$. The altitude splits $\triangle ABC$ into two right triangles: $\triangle ADC$ (with angle $\alpha$ at $A$) and $\triangle BDC$ (with angle $\beta$ at $B$).

### Problem 2 — Two views of the same height $h$

**(a)** In $\triangle ADC$ the side opposite $\alpha$ is $h$, the hypotenuse is $b$:
$$\sin\alpha = \frac{h}{b}$$

**(b)** Solve for $h$:
$$h = b\sin\alpha$$

**(c)** Substitute into $\text{Area} = \tfrac{1}{2}ch$:
$$\text{Area} = \tfrac{1}{2}\,bc\sin\alpha$$

**(d)** From $\triangle BDC$: $\sin\beta = h/a$, so $h = a\sin\beta$, and
$$\text{Area} = \tfrac{1}{2}\,ac\sin\beta$$

**(e)** Drop the altitude $h_A$ from $A$ onto $BC$. In the new right triangle the hypotenuse is $c$ and the angle at $B$ is $\beta$:
$$h_A = c\sin\beta$$

### Problem 3 — Equating heights gives the Law of Sines

**(a)** Two expressions for the same $h$:
$$h = b\sin\alpha \qquad h = a\sin\beta$$

**(b)** Equating them:
$$a\sin\beta = b\sin\alpha \;\Longrightarrow\; \frac{a}{\sin\alpha} = \frac{b}{\sin\beta}$$

**(c)** Looking at $h_A$ from two angles:
$$h_A = c\sin\beta = b\sin\gamma$$

**(d)** Combining gives the **Law of Sines**:
$$\boxed{\;\frac{a}{\sin\alpha} = \frac{b}{\sin\beta} = \frac{c}{\sin\gamma}\;}$$

## Visualization 1 — The figure as drawn in the book

Triangle $ABC$ with altitude $h = CD$. The two right sub-triangles $\triangle ADC$ and $\triangle BDC$ are shaded so you can see where $\sin\alpha = h/b$ and $\sin\beta = h/a$ live.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon, Arc, FancyArrowPatch

# Choose two angles alpha (at A) and beta (at B); gamma follows.
alpha = np.deg2rad(55)
beta  = np.deg2rad(70)
gamma = np.pi - alpha - beta

# Place A at origin, B on the x-axis. Pick c (= AB) = 5.
c = 5.0
A = np.array([0.0, 0.0])
B = np.array([c,   0.0])

# By Law of Sines: a/sin(alpha) = b/sin(beta) = c/sin(gamma)
a = c * np.sin(alpha) / np.sin(gamma)   # side BC, opposite A
b = c * np.sin(beta)  / np.sin(gamma)   # side AC, opposite B

# C is found by going from A at angle alpha, distance b.
C = A + b * np.array([np.cos(alpha), np.sin(alpha)])
# Foot of altitude from C onto AB (which is the x-axis):
D = np.array([C[0], 0.0])
h = C[1]

fig, ax = plt.subplots(figsize=(9, 6))

# Shade the two right sub-triangles
ax.add_patch(Polygon([A, D, C], closed=True, facecolor='#9ec5fe', alpha=0.45, edgecolor='none'))
ax.add_patch(Polygon([D, B, C], closed=True, facecolor='#ffd09e', alpha=0.55, edgecolor='none'))

# Triangle ABC
tri = Polygon([A, B, C], closed=True, fill=False, edgecolor='black', linewidth=2)
ax.add_patch(tri)

# Altitude h (CD) — dashed
ax.plot([C[0], D[0]], [C[1], D[1]], 'k--', linewidth=1.5)

# Right-angle marker at D
s = 0.18
ax.plot([D[0]-s, D[0]-s, D[0]],
        [D[1],   D[1]+s, D[1]+s], 'k-', linewidth=1.2)

# Vertex labels
ax.annotate('A', A, xytext=(-14, -14), textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('B', B, xytext=( 8,  -14), textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('C', C, xytext=(-4,  10),  textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('D', D, xytext=(-4, -18), textcoords='offset points', fontsize=12)

# Side labels
mid = lambda P, Q: (P + Q) / 2
ax.annotate('c', mid(A, B), xytext=(0, -16), textcoords='offset points', fontsize=13, ha='center', style='italic')
ax.annotate('b', mid(A, C), xytext=(-14, 6),  textcoords='offset points', fontsize=13, ha='center', style='italic')
ax.annotate('a', mid(B, C), xytext=( 12, 6),  textcoords='offset points', fontsize=13, ha='center', style='italic')
ax.annotate('h', mid(C, D), xytext=( 8,  0),  textcoords='offset points', fontsize=13, ha='center', style='italic', color='#b00')

# Angle arcs
def angle_arc(center, vec1, vec2, radius, label, label_offset=(0, 0), color='black'):
    theta1 = np.rad2deg(np.arctan2(vec1[1], vec1[0]))
    theta2 = np.rad2deg(np.arctan2(vec2[1], vec2[0]))
    if theta2 < theta1:
        theta1, theta2 = theta2, theta1
    arc = Arc(center, 2*radius, 2*radius, angle=0, theta1=theta1, theta2=theta2,
              color=color, linewidth=1.5)
    ax.add_patch(arc)
    mid_a = np.deg2rad((theta1 + theta2) / 2)
    pos = center + (radius + 0.18) * np.array([np.cos(mid_a), np.sin(mid_a)])
    ax.annotate(label, pos + np.array(label_offset), fontsize=14, ha='center', va='center', color=color)

angle_arc(A, B - A, C - A, 0.55, r'$\alpha$', color='#1f4fa8')
angle_arc(B, A - B, C - B, 0.55, r'$\beta$',  color='#a8541f')
angle_arc(C, A - C, B - C, 0.55, r'$\gamma$', color='#1a7a1a')

ax.set_xlim(-0.8, c + 0.8)
ax.set_ylim(-0.8, h + 0.9)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Triangle ABC with altitude $h = CD$ — Gelfand Ch. 1, Problem 2', fontsize=13)

# Print the numeric witnesses of every formula
print(f'a = {a:.4f},  b = {b:.4f},  c = {c:.4f},  h = {h:.4f}')
print(f'b sin(alpha) = {b*np.sin(alpha):.4f}   (should equal h)')
print(f'a sin(beta)  = {a*np.sin(beta):.4f}   (should equal h)')
print(f'Area via (1/2)ch    = {0.5*c*h:.4f}')
print(f'Area via (1/2)bc sin a = {0.5*b*c*np.sin(alpha):.4f}')
print(f'Area via (1/2)ac sin b = {0.5*a*c*np.sin(beta):.4f}')
print()
print('Law of Sines ratios:')
print(f'  a / sin(alpha) = {a/np.sin(alpha):.4f}')
print(f'  b / sin(beta)  = {b/np.sin(beta):.4f}')
print(f'  c / sin(gamma) = {c/np.sin(gamma):.4f}')

plt.show()
```

## Visualization 2 — Problem 2(e): the second altitude $h_A$ from $A$

This is the "new diagram" the book hints at. Drop a perpendicular from $A$ onto side $BC$. Now $c$ is the hypotenuse and the angle at $B$ is $\beta$, giving $h_A = c\sin\beta$. Equating two views of $h_A$ yields $c\sin\beta = b\sin\gamma$ — Problem 3(c).

```python
# Foot of perpendicular from A onto line BC
BC = C - B
t = np.dot(A - B, BC) / np.dot(BC, BC)
E = B + t * BC                   # foot of altitude from A
hA = np.linalg.norm(A - E)

fig, ax = plt.subplots(figsize=(9, 6))

# Shade the right triangle ABE (angle beta at B, hypotenuse c, opposite hA)
ax.add_patch(Polygon([A, B, E], closed=True, facecolor='#c9e7c9', alpha=0.7, edgecolor='none'))
# And the right triangle ACE (hypotenuse b, angle gamma at C, opposite hA)
ax.add_patch(Polygon([A, C, E], closed=True, facecolor='#f4c4f4', alpha=0.45, edgecolor='none'))

# Triangle ABC
ax.add_patch(Polygon([A, B, C], closed=True, fill=False, edgecolor='black', linewidth=2))

# New altitude h_A (AE) — dashed
ax.plot([A[0], E[0]], [A[1], E[1]], 'r--', linewidth=1.7)

# Right-angle marker at E (perpendicular to BC)
u = (C - B) / np.linalg.norm(C - B)        # along BC
v = (A - E) / np.linalg.norm(A - E)        # along altitude
sq_size = 0.18
p1 = E + sq_size * u
p2 = E + sq_size * u + sq_size * v
p3 = E + sq_size * v
ax.plot([p1[0], p2[0], p3[0]], [p1[1], p2[1], p3[1]], 'k-', linewidth=1.2)

# Labels
ax.annotate('A', A, xytext=(-14, -14), textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('B', B, xytext=( 8,  -14), textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('C', C, xytext=(-4,  10),  textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('E', E, xytext=( 8,   4),  textcoords='offset points', fontsize=12)
ax.annotate(r'$h_A = c\sin\beta = b\sin\gamma$', mid(A, E),
            xytext=(-90, 0), textcoords='offset points', fontsize=12, color='#b00')
ax.annotate('c', mid(A, B), xytext=(0, -16), textcoords='offset points', fontsize=13, ha='center', style='italic')
ax.annotate('b', mid(A, C), xytext=(-14, 6), textcoords='offset points', fontsize=13, ha='center', style='italic')
ax.annotate('a', mid(B, C), xytext=( 12, 6), textcoords='offset points', fontsize=13, ha='center', style='italic')

angle_arc(B, A - B, C - B, 0.55, r'$\beta$',  color='#a8541f')
angle_arc(C, A - C, B - C, 0.55, r'$\gamma$', color='#1a7a1a')

ax.set_xlim(-0.8, c + 0.8)
ax.set_ylim(-0.8, h + 0.9)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Second altitude: from $A$ to side $BC$  (Problem 2e / 3c)', fontsize=13)

print(f'h_A (computed)  = {hA:.4f}')
print(f'c sin(beta)     = {c*np.sin(beta):.4f}')
print(f'b sin(gamma)    = {b*np.sin(gamma):.4f}')
print(f'Area via (1/2) a h_A = {0.5*a*hA:.4f}   (matches the other area formulas)')

plt.show()
```

## Visualization 3 — Interactive proof: drag the angles, watch $a/\sin\alpha$ stay constant

Sweep $\alpha$ and $\beta$ over a range of acute values and plot the three Law-of-Sines ratios. They should sit on top of each other for every triangle — that's the law itself, in pictures.

```python
alphas = np.deg2rad(np.linspace(20, 80, 60))
beta_fixed = np.deg2rad(60)
c_fixed = 1.0

ratios_a, ratios_b, ratios_c = [], [], []
for al in alphas:
    ga = np.pi - al - beta_fixed
    if ga <= 0:
        ratios_a.append(np.nan); ratios_b.append(np.nan); ratios_c.append(np.nan)
        continue
    bb = c_fixed * np.sin(beta_fixed) / np.sin(ga)
    aa = c_fixed * np.sin(al)         / np.sin(ga)
    ratios_a.append(aa / np.sin(al))
    ratios_b.append(bb / np.sin(beta_fixed))
    ratios_c.append(c_fixed / np.sin(ga))

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(np.rad2deg(alphas), ratios_a, label=r'$a/\sin\alpha$', linewidth=2.5)
ax.plot(np.rad2deg(alphas), ratios_b, '--', label=r'$b/\sin\beta$',  linewidth=2.0)
ax.plot(np.rad2deg(alphas), ratios_c, ':',  label=r'$c/\sin\gamma$', linewidth=2.5)
ax.set_xlabel(r'$\alpha$ (degrees), with $\beta = 60^\circ$, $c = 1$')
ax.set_ylabel('ratio')
ax.set_title('All three Law-of-Sines ratios coincide for every acute triangle')
ax.grid(alpha=0.3)
ax.legend(fontsize=12)
plt.show()
```

## What to take away

- The altitude $h$ is the **bridge**: looked at from $A$ it equals $b\sin\alpha$; looked at from $B$ it equals $a\sin\beta$. Equating the two views immediately gives $\dfrac{a}{\sin\alpha} = \dfrac{b}{\sin\beta}$.
- Dropping a *different* altitude ($h_A$ from $A$) gives the third ratio in the same way.
- Three sides, three angles, **one** common ratio — the Law of Sines.
- The area formula $\tfrac{1}{2}bc\sin\alpha$ is just "$\tfrac{1}{2}$ base $\times$ height" with the height rewritten via $\sin$.
