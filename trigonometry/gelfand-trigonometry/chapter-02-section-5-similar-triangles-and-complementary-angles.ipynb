# Chapter 2 · Section 5 — Similar Triangles from an Altitude, and the Complementary-Angle Identity

**Book:** Gelfand & Saul — *Trigonometry*

This notebook is built **intuition first**. Two ideas, both provable by looking at a picture:

1. Dropping the altitude from the right angle of a right triangle creates **three triangles that are exact scale copies of each other**. That gives us three different-looking, secretly-identical formulas for $\cos\alpha$.
2. The two acute angles $\alpha$ and $\beta$ of any right triangle add to $90^\circ$, and what one calls **adjacent** the other calls **opposite**. That single sentence *is* the identity

$$\sin\alpha \;=\; \cos(90^\circ - \alpha).$$

The word *cosine* literally means *complementary sine* — this section is where that name earns its keep.

---

## Part 1 — The Three Nested Copies

### 1.1 The core intuition

Start with a right triangle. Bottom-left angle is $\alpha$, bottom-right is the right angle. Since the three angles sum to $180^\circ$, the top angle must be $90^\circ - \alpha$.

Now **drop the altitude** $d$ from the right angle down to the hypotenuse. That single line splits the big triangle into two smaller right triangles. Here is the crucial observation, done just by chasing angles:

- **Left (medium) triangle.** Its bottom-left angle is still $\alpha$ (we didn't touch that corner). It has a right angle where $d$ meets the hypotenuse. So its remaining angle must be $90^\circ - \alpha$.
- **Right (small) triangle.** Its bottom angle is a right angle. The top corner of the *original* big triangle was $90^\circ - \alpha$, and the left triangle already used up a $90^\circ - \alpha$ piece at the top of $d$... wait, look again. Actually the top vertex of the big triangle is shared with the top vertex of the **small** right triangle. So the small triangle has $90^\circ - \alpha$ at its bottom-right corner, and its top corner must be $\alpha$.

**Key takeaway.** All three triangles have the same three angles $\{\alpha,\; 90^\circ - \alpha,\; 90^\circ\}$. They are **similar**. Only the sizes differ.

### 1.2 What does $\cos\alpha$ mean?

$$\cos\alpha \;=\; \frac{\text{side adjacent to }\alpha}{\text{hypotenuse}}.$$

Because we have three similar right triangles in the picture, we can compute $\cos\alpha$ **three different ways**, and all three must give the same number.

Label the sides:

- $c$ = full hypotenuse of the big triangle
- $a$ = side adjacent to $\alpha$ in the big triangle (bottom side)
- $b$ = the other leg of the big triangle
- $d$ = altitude from the right angle to the hypotenuse
- $p$ = piece of the hypotenuse next to $\alpha$
- $q$ = piece of the hypotenuse away from $\alpha$   (so $p + q = c$)

| Triangle | Angle $\alpha$ sits at | Adjacent leg | Hypotenuse | $\cos\alpha$ |
|---|---|---|---|---|
| **Big** | bottom-left | $a$ | $c$ | $\displaystyle\frac{a}{c}$ |
| **Medium** (left) | bottom-left | $p$ | $a$ | $\displaystyle\frac{p}{a}$ |
| **Small** (right) | **top** vertex | $d$ | $b$ | $\displaystyle\frac{d}{b}$ |

So immediately:

$$\boxed{\;\cos\alpha \;=\; \frac{a}{c} \;=\; \frac{p}{a} \;=\; \frac{d}{b}\;}$$

### 1.3 Visualization — the three similar triangles, colour-coded

The big triangle is outlined in black. Its three interior right triangles are shaded in three different colours so you can see them as separate objects. Every one of them carries the same three angles $\{\alpha,\; 90^\circ-\alpha,\; 90^\circ\}$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon, Arc

# --- Big right triangle: right angle at the top vertex C ---
# Layout: A bottom-left (angle alpha), B bottom-right, C above line AB (right angle at C).
alpha_deg = 35.0
alpha = np.deg2rad(alpha_deg)

c_len = 6.0                          # hypotenuse AB
A = np.array([0.0, 0.0])
B = np.array([c_len, 0.0])

# C is the vertex with the right angle. Distances:
# From A: side b = c cos(alpha), at direction alpha above horizontal
b = c_len * np.cos(alpha)
a = c_len * np.sin(alpha)            # (this will be the OTHER leg AC of the big right triangle)
# Actually let's set it up so alpha is at A, the leg adjacent to alpha is AC ("a" in our table),
# and the hypotenuse is AB. Right angle at C.
# Adjacent to alpha in the BIG triangle has length c*cos(alpha).
adj_big = c_len * np.cos(alpha)      # will be labelled 'a' in the picture
opp_big = c_len * np.sin(alpha)      # will be labelled 'b' in the picture

# Place C so that AC has length adj_big at angle alpha above AB.
C = A + adj_big * np.array([np.cos(alpha), np.sin(alpha)])

# Foot of altitude from C onto AB (x-axis).
D = np.array([C[0], 0.0])
d_len = C[1]                          # altitude length
p_len = D[0]                          # piece of hypotenuse next to alpha
q_len = c_len - p_len                 # piece away from alpha

fig, ax = plt.subplots(figsize=(11, 6.5))

# Medium triangle (left):  A, D, C   -> blue
ax.add_patch(Polygon([A, D, C], closed=True, facecolor='#9ec5fe', alpha=0.55, edgecolor='none'))
# Small triangle (right):  D, B, C   -> orange
ax.add_patch(Polygon([D, B, C], closed=True, facecolor='#ffd09e', alpha=0.65, edgecolor='none'))

# Big triangle outline
ax.add_patch(Polygon([A, B, C], closed=True, fill=False, edgecolor='black', linewidth=2.2))

# Altitude d (dashed)
ax.plot([C[0], D[0]], [C[1], D[1]], 'k--', linewidth=1.6)

# Right-angle square at C (big triangle's right angle)
s = 0.22
u_CA = (A - C) / np.linalg.norm(A - C)
u_CB = (B - C) / np.linalg.norm(B - C)
p1 = C + s * u_CA
p2 = C + s * u_CA + s * u_CB
p3 = C + s * u_CB
ax.plot([p1[0], p2[0], p3[0]], [p1[1], p2[1], p3[1]], 'k-', linewidth=1.2)

# Right-angle square at D (between altitude and hypotenuse)
s2 = 0.18
ax.plot([D[0]-s2, D[0]-s2, D[0]], [D[1], D[1]+s2, D[1]+s2], 'k-', linewidth=1.2)
ax.plot([D[0]+s2, D[0]+s2, D[0]], [D[1], D[1]+s2, D[1]+s2], 'k-', linewidth=1.2)

# Vertex labels
ax.annotate('A', A, xytext=(-14, -14), textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('B', B, xytext=(8,  -14),  textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('C', C, xytext=(-6, 10),   textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('D', D, xytext=(-4, -18),  textcoords='offset points', fontsize=12)

# Side labels
mid = lambda P, Q: (P + Q) / 2
ax.annotate('c',  mid(A, B), xytext=(0, -18), textcoords='offset points', fontsize=13, ha='center', style='italic')
ax.annotate('a',  mid(A, C), xytext=(-16, 8), textcoords='offset points', fontsize=13, ha='center', style='italic', color='#1f4fa8')
ax.annotate('b',  mid(B, C), xytext=( 14, 8), textcoords='offset points', fontsize=13, ha='center', style='italic', color='#a8541f')
ax.annotate('d',  mid(C, D), xytext=( 10, 0), textcoords='offset points', fontsize=13, ha='center', style='italic', color='#b00')
ax.annotate('p',  mid(A, D), xytext=(0, 6),   textcoords='offset points', fontsize=12, ha='center', style='italic', color='#1f4fa8')
ax.annotate('q',  mid(D, B), xytext=(0, 6),   textcoords='offset points', fontsize=12, ha='center', style='italic', color='#a8541f')

# Angle arcs
def angle_arc(center, vec1, vec2, radius, label, color='black', label_shift=(0, 0)):
    t1 = np.rad2deg(np.arctan2(vec1[1], vec1[0]))
    t2 = np.rad2deg(np.arctan2(vec2[1], vec2[0]))
    if t2 < t1: t1, t2 = t2, t1
    arc = Arc(center, 2*radius, 2*radius, angle=0, theta1=t1, theta2=t2, color=color, linewidth=1.6)
    ax.add_patch(arc)
    mid_a = np.deg2rad((t1 + t2) / 2)
    pos = center + (radius + 0.28) * np.array([np.cos(mid_a), np.sin(mid_a)])
    ax.annotate(label, pos + np.array(label_shift), fontsize=13, ha='center', va='center', color=color)

# alpha at A
angle_arc(A, B - A, C - A, 0.7, r'$\alpha$', color='#1f4fa8')
# 90 - alpha at B
angle_arc(B, A - B, C - B, 0.7, r'$90^\circ\!-\!\alpha$', color='#a8541f')
# At C the two sub-angles: alpha on the right side of d (towards B), 90-alpha on the left (towards A)
angle_arc(C, D - C, B - C, 0.6, r'$\alpha$', color='#a8541f')
angle_arc(C, A - C, D - C, 0.6, r'$90^\circ\!-\!\alpha$', color='#1f4fa8')

ax.set_xlim(-0.9, c_len + 0.9)
ax.set_ylim(-0.9, d_len + 1.2)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Three similar right triangles: big (outline), medium (blue), small (orange)', fontsize=13)

# Numeric witnesses
print(f'alpha            = {alpha_deg}°')
print(f'a (adj to alpha) = {adj_big:.4f}    c (hyp) = {c_len:.4f}')
print(f'b (opp to alpha) = {opp_big:.4f}    d       = {d_len:.4f}')
print(f'p                = {p_len:.4f}    q       = {q_len:.4f}')
print()
print('Three views of cos(alpha):')
print(f'  big:    a/c = {adj_big/c_len:.6f}')
print(f'  medium: p/a = {p_len/adj_big:.6f}')
print(f'  small:  d/b = {d_len/opp_big:.6f}')
print(f'  cos({alpha_deg}°)  = {np.cos(alpha):.6f}   <-- all four numbers should agree')

plt.show()
```

### 1.4 Zoom in on the small triangle — why does $\alpha$ appear at the *top* corner?

Students almost always trip on the third formula $\cos\alpha = d/b$. The reason is that $\alpha$ isn't at the bottom of the small triangle — it moved to the top. Here's the two-step chase again, isolated:

1. **The top of the big triangle is a right angle** (that's where we put the right angle in the setup). Actually — here we set the right angle at $C$, at the top. The altitude $d$ splits that $90^\circ$ into two pieces: $90^\circ - \alpha$ on the left (part of the medium triangle) and $\alpha$ on the right (part of the small triangle).
2. **Sanity check inside the small triangle.** Its angles must sum to $180^\circ$. It has a right angle at $D$ and $\alpha$ at $C$. The remaining angle at $B$ is $180^\circ - 90^\circ - \alpha = 90^\circ - \alpha$. ✓ Matches the big triangle's angle at $B$.

Now sit at that top corner $C$ where $\alpha$ lives, in the small triangle only:

- **Adjacent** to $\alpha$: the altitude $d$ (the leg that touches $\alpha$ and is not the hypotenuse).
- **Opposite** $\alpha$: the small piece $q$ at the bottom.
- **Hypotenuse**: the outer slanted side $b$ (the side facing the right angle at $D$).

So $\displaystyle \cos\alpha = \frac{d}{b}$.

The visualization below strips away the big triangle and shows only the small right triangle, with $\alpha$ highlighted at the top.

```python
# Isolate the small right triangle DBC and re-draw it in a clean frame.
fig, ax = plt.subplots(figsize=(7, 6))

# Shift vertices so D is at origin for a clean picture
shift = D.copy()
Dp = D - shift
Bp = B - shift
Cp = C - shift

ax.add_patch(Polygon([Dp, Bp, Cp], closed=True, facecolor='#ffd09e', alpha=0.6, edgecolor='black', linewidth=2))

# Right-angle square at D
s2 = 0.15
ax.plot([Dp[0]+s2, Dp[0]+s2, Dp[0]], [Dp[1], Dp[1]+s2, Dp[1]+s2], 'k-', linewidth=1.2)

# Vertex labels
ax.annotate('D',  Dp, xytext=(-14, -14), textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('B',  Bp, xytext=(8,   -14), textcoords='offset points', fontsize=14, fontweight='bold')
ax.annotate('C',  Cp, xytext=(-6,  12),  textcoords='offset points', fontsize=14, fontweight='bold')

# Side labels
ax.annotate('q (opposite $\\alpha$)', mid(Dp, Bp), xytext=(0, -22), textcoords='offset points',
            fontsize=12, ha='center', style='italic')
ax.annotate('d (adjacent to $\\alpha$)', mid(Cp, Dp), xytext=(-92, 0), textcoords='offset points',
            fontsize=12, ha='center', style='italic', color='#b00')
ax.annotate('b (hypotenuse)', mid(Bp, Cp), xytext=(14, 8), textcoords='offset points',
            fontsize=12, ha='center', style='italic', color='#a8541f')

# alpha at top vertex C, between edges CD and CB
angle_arc(Cp, Dp - Cp, Bp - Cp, 0.6, r'$\alpha$', color='#1f4fa8')
# 90 - alpha at B
angle_arc(Bp, Dp - Bp, Cp - Bp, 0.6, r'$90^\circ\!-\!\alpha$', color='#a8541f')

ax.set_xlim(Dp[0] - 1.0, Bp[0] + 1.6)
ax.set_ylim(Dp[1] - 0.9, Cp[1] + 1.0)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title(r'Small triangle only — $\alpha$ lives at the TOP vertex $C$', fontsize=13)

print('Small triangle sides:')
print(f'  d (adjacent to alpha) = {d_len:.4f}')
print(f'  q (opposite alpha)    = {q_len:.4f}')
print(f'  b (hypotenuse)        = {opp_big:.4f}')
print()
print(f'  cos(alpha) = d / b = {d_len/opp_big:.6f}')
print(f'  sin(alpha) = q / b = {q_len/opp_big:.6f}')
print(f'  Ground truth: cos({alpha_deg}°) = {np.cos(alpha):.6f},  sin({alpha_deg}°) = {np.sin(alpha):.6f}')

plt.show()
```

### 1.5 Bonus: two classical identities fall out for free

The three-similar-triangles picture is quietly powerful. From $\cos\alpha = a/c = p/a$, cross-multiply:

$$a^2 \;=\; c\,p.$$

By the same reasoning applied to the small triangle, $b^2 = c\,q$. Adding these:

$$a^2 + b^2 \;=\; c\,(p + q) \;=\; c\cdot c \;=\; c^2.$$

That's the **Pythagorean theorem**, produced by dropping one altitude and reading off similar triangles — no rearrangement of squares needed.

And from the medium and small triangles, $\sin\alpha = d/a$ and $\cos\alpha = d/b$ give

$$d \;=\; a\sin\alpha \;=\; b\cos\alpha.$$

One line, two facts.

## Part 2 — The Complementary-Angle Identity

$$\sin\alpha \;=\; \cos(90^\circ - \alpha)$$

### 2.1 The core intuition — two people, one triangle

Take any right triangle. Call the two acute angles $\alpha$ (bottom-left) and $\beta$ (top). Two students sit inside:

- **Person $A$** is at corner $\alpha$.
- **Person $B$** is at corner $\beta$.

Both are staring at the same short bottom leg. Ask each of them what role that leg plays *from where they are sitting*:

- Person $A$ says: “That leg is right next to me — it's my **adjacent** side.” So for them, $\cos\alpha = \dfrac{\text{that leg}}{\text{hypotenuse}}$.
- Person $B$ says: “That leg is directly across from me — it's my **opposite** side.” So for them, $\sin\beta = \dfrac{\text{that leg}}{\text{hypotenuse}}$.

**They are measuring the same fraction.** Same numerator, same denominator. Therefore

$$\cos\alpha \;=\; \sin\beta.$$

### 2.2 Why $\alpha + \beta = 90^\circ$

The angles of a triangle sum to $180^\circ$. One of them is $90^\circ$. So the other two must sum to $90^\circ$:

$$\alpha + \beta = 90^\circ \;\Longleftrightarrow\; \beta = 90^\circ - \alpha.$$

Substitute $\beta = 90^\circ - \alpha$ into $\cos\alpha = \sin\beta$:

$$\boxed{\;\sin(90^\circ - \alpha) = \cos\alpha,\qquad \cos(90^\circ - \alpha) = \sin\alpha.\;}$$

That is exactly why $\cos$ is called the *cosine* — the **co**mplementary **sine**.

### 2.3 Visualization — the same leg, two viewpoints

The shared leg is drawn in red. From $A$'s corner it is the **adjacent** side of angle $\alpha$. From $B$'s corner it is the **opposite** side of angle $\beta$. Same red segment, same hypotenuse — so

$$\cos\alpha = \frac{\text{red}}{\text{hyp}} = \sin\beta.$$

```python
# Right triangle with alpha at A (bottom-left), right angle at C (bottom-right), beta at B (top).
alpha2_deg = 35.0
alpha2 = np.deg2rad(alpha2_deg)
beta2  = np.pi/2 - alpha2
beta2_deg = 90 - alpha2_deg

hyp = 5.0
adj = hyp * np.cos(alpha2)   # bottom leg, adjacent to alpha, opposite to beta
opp = hyp * np.sin(alpha2)   # vertical leg

A2 = np.array([0.0, 0.0])
C2 = np.array([adj, 0.0])
B2 = np.array([adj, opp])

fig, ax = plt.subplots(figsize=(9, 6))

# Triangle
ax.add_patch(Polygon([A2, C2, B2], closed=True, facecolor='#eaf4ff', edgecolor='black', linewidth=2))

# Highlight the shared bottom leg in red
ax.plot([A2[0], C2[0]], [A2[1], C2[1]], color='#b00', linewidth=4.5, solid_capstyle='butt')

# Right-angle square at C
s = 0.18
ax.plot([C2[0]-s, C2[0]-s, C2[0]], [C2[1], C2[1]+s, C2[1]+s], 'k-', linewidth=1.2)

# Vertex labels
ax.annotate('A (Person A sits here)', A2, xytext=(-6, -20), textcoords='offset points', fontsize=12, fontweight='bold', ha='left')
ax.annotate('C', C2, xytext=(8, -18),  textcoords='offset points', fontsize=13, fontweight='bold')
ax.annotate('B (Person B sits here)', B2, xytext=(8, 8), textcoords='offset points', fontsize=12, fontweight='bold')

# Side labels
ax.annotate(f'adj to α  /  opp to β\n(shared red side)', mid(A2, C2),
            xytext=(0, -34), textcoords='offset points', fontsize=11, ha='center', color='#b00')
ax.annotate('hypotenuse', mid(A2, B2), xytext=(-24, 10), textcoords='offset points',
            fontsize=11, ha='center', rotation=np.rad2deg(alpha2))
ax.annotate('opposite α / adjacent β', mid(C2, B2), xytext=(14, 0), textcoords='offset points',
            fontsize=11, ha='left')

# Angle arcs
angle_arc(A2, C2 - A2, B2 - A2, 0.55, rf'$\alpha={alpha2_deg:.0f}^\circ$', color='#1f4fa8')
angle_arc(B2, A2 - B2, C2 - B2, 0.55, rf'$\beta={beta2_deg:.0f}^\circ$',  color='#a8541f')

ax.set_xlim(-1.2, adj + 2.5)
ax.set_ylim(-1.0, opp + 1.0)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title(r'Same red side: $\cos\alpha = \text{red}/\text{hyp} = \sin\beta$', fontsize=13)

print(f'alpha = {alpha2_deg}°,   beta = {beta2_deg}°,   alpha + beta = {alpha2_deg + beta2_deg}°')
print()
print(f'red side (bottom leg) = {adj:.4f}')
print(f'hypotenuse            = {hyp:.4f}')
print()
print(f'cos(alpha) = adj/hyp = {adj/hyp:.6f}')
print(f'sin(beta)  = opp/hyp = {adj/hyp:.6f}   (same number by construction)')
print(f'Numpy check: cos({alpha2_deg}°) = {np.cos(alpha2):.6f},   sin({beta2_deg}°) = {np.sin(beta2):.6f}')

plt.show()
```

### 2.4 Numerical witness — sweep $\alpha$ from $0^\circ$ to $90^\circ$

Plot $\sin\alpha$ and $\cos(90^\circ - \alpha)$ on top of each other. If the identity is real, they must be indistinguishable curves.

```python
alphas = np.linspace(0, 90, 200)
sin_alpha = np.sin(np.deg2rad(alphas))
cos_comp  = np.cos(np.deg2rad(90 - alphas))

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(alphas, sin_alpha, label=r'$\sin\alpha$', linewidth=3)
ax.plot(alphas, cos_comp, '--', label=r'$\cos(90^\circ-\alpha)$', linewidth=2)
ax.set_xlabel(r'$\alpha$ (degrees)')
ax.set_ylabel('value')
ax.set_title(r'$\sin\alpha = \cos(90^\circ - \alpha)$ — one curve, drawn twice')
ax.grid(alpha=0.3)
ax.legend(fontsize=12)

max_diff = np.max(np.abs(sin_alpha - cos_comp))
print(f'max |sin(alpha) - cos(90 - alpha)| over 0..90 = {max_diff:.2e}')

plt.show()
```

## Part 3 — Worked exercises

### Exercise 1 — Show that $\sin 29^\circ = \cos 61^\circ$

Draw a right triangle with an acute angle of $29^\circ$. The other acute angle is $90^\circ - 29^\circ = 61^\circ$.

The side **opposite** the $29^\circ$ corner is the exact same segment that sits **adjacent** to the $61^\circ$ corner (that is what the two-person argument in 2.1 shows). Divide by the shared hypotenuse:

$$\sin 29^\circ \;=\; \frac{\text{that side}}{\text{hyp}} \;=\; \cos 61^\circ.\qquad\blacksquare$$

### Exercise 2 — If $\sin 35^\circ = \cos x$, find $x$

For sine of one angle to equal cosine of another, the two angles must be the two acute corners of a single right triangle — so they are complementary:

$$35^\circ + x = 90^\circ \;\Longrightarrow\; x = 55^\circ.$$

### Exercise 3 — Rewrite $\sin\alpha = \cos\beta$ using only $\alpha$

In the right triangle picture, $\beta = 90^\circ - \alpha$. Substitute:

$$\sin\alpha \;=\; \cos(90^\circ - \alpha).$$

That is the identity in its cleanest form.

```python
# Numerical sanity checks for the three exercises.
print('Exercise 1:  sin 29°  vs  cos 61°')
print(f'  sin(29°) = {np.sin(np.deg2rad(29)):.10f}')
print(f'  cos(61°) = {np.cos(np.deg2rad(61)):.10f}')
print()

print('Exercise 2:  sin 35° = cos x  =>  x = 55°')
print(f'  sin(35°) = {np.sin(np.deg2rad(35)):.10f}')
print(f'  cos(55°) = {np.cos(np.deg2rad(55)):.10f}')
print()

print('Exercise 3:  sin(alpha) = cos(90° - alpha), spot-checked at several alphas')
for a in [10, 23.7, 45, 62, 78.5]:
    lhs = np.sin(np.deg2rad(a))
    rhs = np.cos(np.deg2rad(90 - a))
    print(f'  alpha = {a:5.1f}°:  sin = {lhs:.10f},  cos(90-α) = {rhs:.10f},  diff = {abs(lhs-rhs):.1e}')
```

## What to take away

- **One line does most of the work.** In a right triangle, drop the altitude from the right angle. You instantly get **three similar right triangles**, and any trig ratio you can write about $\alpha$ using the big triangle can be re-written two more ways using the smaller ones.
- **Adjacent for one angle = opposite for its complement.** That's the whole content of $\sin\alpha = \cos(90^\circ - \alpha)$. No memorization needed — just the two-person picture.
- The Pythagorean theorem, the identity $d = a\sin\alpha = b\cos\alpha$, and every complementary-angle identity all live inside the *same* picture.
