## Exercise 2.156: Accidental Deaths by Age and Type

### The Problem

The table below lists accidental deaths by age and certain specific types for the United States in 2002.

| Age | All Types | Motor Vehicle | Falls | Drowning |
|-----|----------:|-------------:|------:|---------:|
| Under 5    | 2,707  | 819    | 44    | 568 |
| 5–14       | 2,979  | 1,772  | 37    | 375 |
| 15–24      | 14,113 | 10,560 | 237   | 646 |
| 25–34      | 11,769 | 6,884  | 303   | 419 |
| 35–44      | 15,413 | 6,927  | 608   | 480 |
| 45–54      | 12,278 | 5,361  | 871   | 354 |
| 55–64      | 7,505  | 3,506  | 949   | 217 |
| 65–74      | 7,698  | 3,038  | 1,660 | 179 |
| 75 and over| 23,438 | 4,487  | 8,613 | 244 |
| **Total**  | **97,900** | **43,354** | **13,322** | **3,482** |

> Source: National Vital Statistics Report 50, no. 15, 2002

---

### Critical Observation: Is "All Types" a Separate Category?

**No.** "All Types" is the *row total*, not a fourth accident type.  
Check: $819 + 44 + 568 = 1{,}431 \ne 2{,}707$ for the Under-5 row — the three listed types cover only part of each total.  
The remaining deaths come from unlisted types (poisoning, fires, choking, etc.).  
Therefore **"All Types" is the sample-space denominator**, not a term to add alongside the specific types.

---

### Part (a): Probability Calculations

A randomly selected person from the United States was known to have had an accidental death in 2002. Find the probability that:

**(i)** The person was over the age of 15.  
**(ii)** The cause of death was a motor vehicle accident.  
**(iii)** The cause was a motor vehicle accident, given the person was between 15 and 24 years old.  
**(iv)** The cause was drowning, given it was not a motor vehicle accident and the person was 34 years old or younger.

### Part (b)

From these figures, can you determine the probability that a person selected at random from the U.S. population had a fatal motor vehicle accident in 2002?

---

### Solutions

**Core principle:** $P(\text{event}) = \dfrac{\text{favorable outcomes}}{\text{total pool}}$. A "given" clause shrinks the pool.

#### (a-i) Person was over 15

Exclude Under 5 and 5–14:

$$P(\text{over 15}) = \frac{97{,}900 - (2{,}707 + 2{,}979)}{97{,}900} = \frac{92{,}214}{97{,}900} \approx 0.9419$$

#### (a-ii) Motor vehicle accident

$$P(\text{MV}) = \frac{43{,}354}{97{,}900} \approx 0.4428$$

#### (a-iii) Motor vehicle, given age 15–24

The condition "given 15–24" restricts the entire world to that row (14,113 deaths):

$$P(\text{MV} \mid 15\text{–}24) = \frac{10{,}560}{14{,}113} \approx 0.7482$$

#### (a-iv) Drowning, given not MV and age $\le 34$

Two simultaneous restrictions — age $\le 34$ (rows: Under 5, 5–14, 15–24, 25–34) and cause $\ne$ Motor Vehicle.

| Step | Calculation | Result |
|------|-------------|-------:|
| Total deaths, age $\le 34$ | $2{,}707+2{,}979+14{,}113+11{,}769$ | $31{,}568$ |
| MV deaths, age $\le 34$ | $819+1{,}772+10{,}560+6{,}884$ | $20{,}035$ |
| **Restricted pool** (not MV, $\le 34$) | $31{,}568 - 20{,}035$ | $11{,}533$ |
| Drowning deaths, age $\le 34$ | $568+375+646+419$ | $2{,}008$ |

$$P(\text{Drowning} \mid \text{not MV},\, \text{age} \le 34) = \frac{2{,}008}{11{,}533} \approx 0.1741$$

#### (b) Can we find $P(\text{fatal MV} \mid \text{random U.S. resident})$?

**No.** Every probability above is *conditional on already having an accidental death*. To compute the probability for any random U.S. resident, we would need the total 2002 U.S. population as the denominator — information not provided in the table.

```python
# Exercise 2.156: Accidental Deaths — numerical verification

data = {
    'Under 5': {'all': 2707,  'mv': 819,   'falls': 44,   'drowning': 568},
    '5-14':    {'all': 2979,  'mv': 1772,  'falls': 37,   'drowning': 375},
    '15-24':   {'all': 14113, 'mv': 10560, 'falls': 237,  'drowning': 646},
    '25-34':   {'all': 11769, 'mv': 6884,  'falls': 303,  'drowning': 419},
    '35-44':   {'all': 15413, 'mv': 6927,  'falls': 608,  'drowning': 480},
    '45-54':   {'all': 12278, 'mv': 5361,  'falls': 871,  'drowning': 354},
    '55-64':   {'all': 7505,  'mv': 3506,  'falls': 949,  'drowning': 217},
    '65-74':   {'all': 7698,  'mv': 3038,  'falls': 1660, 'drowning': 179},
    '75+':     {'all': 23438, 'mv': 4487,  'falls': 8613, 'drowning': 244},
}

total_all = sum(v['all'] for v in data.values())
total_mv  = sum(v['mv']  for v in data.values())

print("Exercise 2.156: Accidental Deaths")
print("=" * 55)

# (a-i) Over 15
under_15 = data['Under 5']['all'] + data['5-14']['all']
over_15  = total_all - under_15
p_i      = over_15 / total_all
print(f"\n(a-i)   P(over 15)         = {over_15:,}/{total_all:,} = {p_i:.4f}")

# (a-ii) Motor vehicle
p_ii = total_mv / total_all
print(f"(a-ii)  P(MV)               = {total_mv:,}/{total_all:,} = {p_ii:.4f}")

# (a-iii) MV | 15-24
row = data['15-24']
p_iii = row['mv'] / row['all']
print(f"(a-iii) P(MV | 15-24)       = {row['mv']:,}/{row['all']:,} = {p_iii:.4f}")

# (a-iv) Drowning | not MV and age <= 34
young_rows  = ['Under 5', '5-14', '15-24', '25-34']
total_young = sum(data[r]['all']      for r in young_rows)
mv_young    = sum(data[r]['mv']       for r in young_rows)
drown_young = sum(data[r]['drowning'] for r in young_rows)
pool_iv     = total_young - mv_young
p_iv        = drown_young / pool_iv
print(f"\n(a-iv)  Restricted pool (not MV, ≤34) = {total_young:,} - {mv_young:,} = {pool_iv:,}")
print(f"        Drowning deaths (≤34)          = {drown_young:,}")
print(f"        P(Drowning | not MV, ≤34)      = {drown_young}/{pool_iv} = {p_iv:.4f}")
```

## Exercise 2.157: Lung Cancer and Smoking

### The Problem

A study of residents in a region showed:
- 20% were smokers: $P(S) = 0.20$, so $P(S^c) = 0.80$.
- A smoker's probability of dying from lung cancer is **ten times** that of a non-smoker:
  $P(L \mid S) = 10 \cdot P(L \mid S^c)$.
- The overall probability of death due to lung cancer: $P(L) = 0.006$.

Find $P(L \mid S)$.

---

### My Approach: Law of Total Probability

#### Step 1: Parameterise

Let $x = P(L \mid S^c)$.  Then $P(L \mid S) = 10x$.

#### Step 2: Law of Total Probability

The overall cancer rate is a weighted average of the two groups:

$$P(L) = P(L \mid S)\,P(S) + P(L \mid S^c)\,P(S^c)$$

$$0.006 = (10x)(0.20) + (x)(0.80)$$

#### Step 3: Solve

$$0.006 = 2.0x + 0.8x = 2.8x \implies x = \frac{0.006}{2.8}$$

#### Step 4: Smoker Rate

$$P(L \mid S) = 10x = \frac{0.06}{2.8} = \frac{3}{140} \approx 0.0214$$

$$\boxed{P(L \mid S) \approx 0.0214 \quad (2.14\%)}$$

#### Intuition Check

Non-smoker rate $x \approx 0.00214$ (0.21%).  
Smoker rate $= 10x \approx 0.0214$ (2.14%) — ten times higher, as required.  
Weighted average: $0.0214 \times 0.20 + 0.00214 \times 0.80 = 0.00428 + 0.00171 = 0.006$ ✓

```python
# Exercise 2.157: Lung Cancer and Smoking

P_S  = 0.20
P_Sc = 0.80
P_L  = 0.006
# P(L|S) = 10 * P(L|Sc)  =>  0.006 = 10x*0.20 + x*0.80  =>  x = 0.006/2.8

x            = P_L / (10 * P_S + P_Sc)
P_L_Sc       = x
P_L_S        = 10 * x

print("Exercise 2.157: Lung Cancer & Smoking")
print("=" * 45)
print(f"P(L | S')  (non-smoker) = {P_L_Sc:.6f}  (~{P_L_Sc*100:.4f}%)")
print(f"P(L | S)   (smoker)     = {P_L_S:.6f}  (~{P_L_S*100:.4f}%)")
print(f"Exact fraction          = {Fraction(P_L_S).limit_denominator(1000)}")
print()

# Verify via total probability
P_L_check = P_L_S * P_S + P_L_Sc * P_Sc
print("Verification — Law of Total Probability:")
print(f"  P(L) = P(L|S)·P(S) + P(L|S')·P(S')")
print(f"       = {P_L_S:.6f}×{P_S} + {P_L_Sc:.6f}×{P_Sc}")
print(f"       = {P_L_check:.6f}  (target: {P_L})")
print(f"  Match: {np.isclose(P_L_check, P_L)}")
```

## Exercise 2.158: Pólya's Urn Scheme

### The Problem

A bowl contains $w$ white balls and $b$ black balls. One ball is drawn at random, its colour noted, and it is returned to the bowl **along with $n$ additional balls of the same colour**. A second ball is then drawn and is observed to be **black**.  
Show that:

$$P(W_1 \mid B_2) = \frac{w}{w + b + n}$$

---

### Proof via Bayes' Theorem

#### Events

| Symbol | Meaning |
|--------|---------|
| $W_1$ | First ball drawn is white |
| $B_1$ | First ball drawn is black |
| $B_2$ | Second ball drawn is black |

#### Step 1: Priors for the First Draw

$$P(W_1) = \frac{w}{w+b}, \qquad P(B_1) = \frac{b}{w+b}$$

#### Step 2: Likelihoods for the Second Draw

**If first was white** ($W_1$): bowl now has $w+n$ white, $b$ black — total $w+b+n$.

$$P(B_2 \mid W_1) = \frac{b}{w+b+n}$$

**If first was black** ($B_1$): bowl now has $w$ white, $b+n$ black — total $w+b+n$.

$$P(B_2 \mid B_1) = \frac{b+n}{w+b+n}$$

#### Step 3: Total Probability of $B_2$

$$P(B_2) = P(B_2 \mid W_1)\,P(W_1) + P(B_2 \mid B_1)\,P(B_1)$$

$$= \frac{b}{w+b+n}\cdot\frac{w}{w+b} + \frac{b+n}{w+b+n}\cdot\frac{b}{w+b}$$

$$= \frac{bw + b(b+n)}{(w+b)(w+b+n)} = \frac{b(w+b+n)}{(w+b)(w+b+n)} = \frac{b}{w+b}$$

> **Beautiful result:** $P(B_2) = P(B_1)$.  
> Pólya's urn is *exchangeable* — the marginal probability of a black ball is the same on every draw, regardless of history.

#### Step 4: Apply Bayes' Theorem

$$P(W_1 \mid B_2) = \frac{P(B_2 \mid W_1)\,P(W_1)}{P(B_2)}
= \frac{\dfrac{b}{w+b+n}\cdot\dfrac{w}{w+b}}{\dfrac{b}{w+b}}$$

The $(w+b)$ factors and $b$ cancel:

$$\boxed{P(W_1 \mid B_2) = \frac{w}{w+b+n}}$$

#### Intuition

Knowing the second ball was black tells us that $n$ extra black balls were added after the first draw. The total pool is now $w+b+n$, and white balls remain $w$ — so the posterior probability mirrors the fraction white would occupy in the *expanded* urn.

```python
# Exercise 2.158: Pólya's Urn — Monte Carlo verification

def polya_sim(w, b, n, n_trials=600_000, seed=42):
    """Estimate P(W1 | B2) by simulation."""
    rng = np.random.default_rng(seed)
    w1_and_b2 = 0
    b2_count  = 0
    for _ in range(n_trials):
        # First draw from {w white, b black}
        first_white = rng.random() < w / (w + b)
        # Update urn
        new_w = w + n if first_white else w
        new_b = b     if first_white else b + n
        # Second draw
        second_black = rng.random() < new_b / (new_w + new_b)
        if second_black:
            b2_count += 1
            if first_white:
                w1_and_b2 += 1
    return w1_and_b2 / b2_count if b2_count > 0 else float('nan')


print("Exercise 2.158: Pólya's Urn")
print("=" * 50)
print(f"{'w':>4} {'b':>4} {'n':>4} | {'Exact P(W1|B2)':>16} | {'Simulated':>12} | Match")
print("-" * 60)

test_cases = [(3, 5, 2), (1, 1, 1), (4, 6, 3), (10, 2, 5)]
for (w, b, n) in test_cases:
    exact = w / (w + b + n)
    sim   = polya_sim(w, b, n)
    match = abs(exact - sim) < 0.005
    print(f"{w:>4} {b:>4} {n:>4} | {exact:>16.6f} | {sim:>12.6f} | {match}")
```

## Exercise 2.159: Show $P(\emptyset) = 0$ from the Axioms

### The Problem

It seems obvious that $P(\emptyset) = 0$. Show that this result *follows from* the axioms in Definition 2.6.

---

### The Axioms (Definition 2.6)

| # | Axiom | Statement |
|---|-------|-----------|
| 1 | Non-negativity | $P(E) \ge 0$ for every event $E$ |
| 2 | Normalization  | $P(S) = 1$ |
| 3 | Countable Additivity | If $E_1, E_2, \ldots$ are pairwise disjoint, then $P\!\left(\bigcup_{i=1}^{\infty} E_i\right) = \sum_{i=1}^{\infty} P(E_i)$ |

---

### Proof

**Step 1: Construct a disjoint sequence**

Define events $E_1 = S$ and $E_i = \emptyset$ for all $i \ge 2$.  
These are pairwise disjoint because $S \cap \emptyset = \emptyset$.

**Step 2: Apply Axioms 2 and 3**

Their union is:
$$S \cup \emptyset \cup \emptyset \cup \cdots = S$$

By Axiom 2, $P(S) = 1$.  
By Axiom 3 (countable additivity):

$$1 = P(S) = P(S) + P(\emptyset) + P(\emptyset) + \cdots = 1 + \sum_{i=2}^{\infty} P(\emptyset)$$

**Step 3: Conclude**

$$\sum_{i=2}^{\infty} P(\emptyset) = 0$$

By Axiom 1, $P(\emptyset) \ge 0$.  
A sum of non-negative terms equals zero *if and only if* every term is zero.  
Therefore:

$$\boxed{P(\emptyset) = 0}$$

#### Why Axiom 1 is essential

Without non-negativity, the series $\sum P(\emptyset) = 0$ could be satisfied by alternating positive and negative terms. Axiom 1 closes that loophole: once we know every term $\ge 0$ and their sum $= 0$, each term must be exactly $0$.

#### Conceptual parallel

The empty event $\emptyset$ contains no outcomes, so it should carry zero probability. The proof makes this intuition rigorous from first principles without appealing to geometric intuition.
