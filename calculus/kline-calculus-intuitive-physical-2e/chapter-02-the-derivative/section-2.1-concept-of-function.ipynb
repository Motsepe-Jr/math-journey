# Kline — Calculus: An Intuitive and Physical Approach (2e)

## Chapter 2: The Derivative

### Section 1 — The Concept of Function

---

Kline opens Chapter 2 by reviewing the idea of a **function** before defining the derivative. The point is that calculus studies how quantities *depend* on one another, so before we can talk about rate of change we need to be crisp about what a functional relationship is.

**Key ideas Kline emphasizes in this section:**

- A function is a rule associating each value of an **independent variable** with a value (or values) of a **dependent variable**.
- The formula tells us which variable is being solved *for* — that's the dependent one.
- A function is **single-valued** if each value of the independent variable produces exactly one value of the dependent variable.
- When we invert (solve for the other variable), we often *lose* single-valuedness — e.g. $y = x^2$ inverted is $x = \pm\sqrt{y}$.
- Physical problems usually pick the physically meaningful branch (a radius is nonnegative, a length is nonnegative, etc.).

Below I work through Exercises 1–8 of the section.

## Exercise 1

> The amount $A$ of money that accumulates in $n$ years if one dollar is invested and if the interest is compounded annually at the fixed rate of $i$ per cent per year is
>
> $$A = (1 + i)^n.$$
>
> As the formula is written, which is the independent variable? Which is the dependent variable?

### My answer

The phrase "at the **fixed** rate $i$" tells us $i$ is a **parameter** — a constant baked into the problem, not a variable we're varying.

So we're left with two quantities that actually move: $n$ (the input we choose) and $A$ (the output the formula produces once $n$ is chosen).

- **Independent variable:** $n$ — the number of years. We pick it.
- **Dependent variable:** $A$ — the accumulated amount. It's determined by our choice of $n$.

The formula is written with $A$ isolated on the left — a standard tip-off that $A$ is the dependent quantity.

## Exercise 2

> Solve $v = 32t$ for $t$. Is the resulting function single-valued?

### My answer

Divide both sides by 32:

$$t = \frac{v}{32}.$$

**Single-valued?** Yes. For any value of $v$, dividing by 32 gives one, and only one, value of $t$. Linear rescalings never introduce multi-valuedness.

(Physically: $v = 32t$ is the speed of a freely falling body $t$ seconds after release, in ft/s. Solving for $t$ says "given the current speed, how long has it been falling?" — a unique answer for each speed.)

## Exercise 3

> Solve $y = 5.3x^2$ for $x$. Is the resulting function single-valued?

### My answer

$$x^2 = \frac{y}{5.3} \quad\Longrightarrow\quad x = \pm\sqrt{\tfrac{y}{5.3}}.$$

**Single-valued?** **No.** For any $y > 0$ there are *two* values of $x$ (one positive, one negative) that both satisfy the equation. This is the classic price of inverting an even-power function.

If we wanted a single-valued function, we'd have to choose a branch — say $x = +\sqrt{y/5.3}$ or $x = -\sqrt{y/5.3}$ — and stick to it.

## Exercise 4

> Write the formula for:
> (a) the area $A$ of a circle in terms of the radius;
> (b) the area $A$ of a circle in terms of the diameter.

### My answer

**(a)** In terms of the radius $r$:

$$A = \pi r^2.$$

**(b)** The diameter is $d = 2r$, so $r = d/2$. Substituting:

$$A = \pi \left(\frac{d}{2}\right)^2 = \frac{\pi d^2}{4}.$$

Both are single-valued functions of their input, and both give the same $A$ when the two inputs correspond to the same circle.

## Exercise 5

> Write the formula for the radius of a circle in terms of the area. Is the function single-valued? If not, which of the single-valued functions do you think would be more useful and why?

### My answer

Solving $A = \pi r^2$ for $r$:

$$r^2 = \frac{A}{\pi} \quad\Longrightarrow\quad r = \pm\sqrt{\tfrac{A}{\pi}}.$$

**Single-valued?** No — the algebra gives two branches, $r = +\sqrt{A/\pi}$ and $r = -\sqrt{A/\pi}$.

**Which branch is more useful?** The **positive** branch,

$$r = +\sqrt{\tfrac{A}{\pi}},$$

because a radius is a *length*, and lengths are nonnegative by definition. The negative branch is algebraically valid but geometrically meaningless — no circle has radius $-3$. This is a good example of Kline's larger point: the *physics* of a problem picks the branch, not the algebra.

## Exercise 6

> One arm of a right triangle is 3 units and the hypotenuse is $x$ units. Write a formula for the length of the other arm. *Suggestion: Use the Pythagorean theorem.*

### My answer

Let the other arm have length $b$. By the Pythagorean theorem,

$$3^2 + b^2 = x^2 \quad\Longrightarrow\quad b^2 = x^2 - 9 \quad\Longrightarrow\quad b = \sqrt{x^2 - 9}.$$

I take the positive square root because $b$ is a length. The formula is only meaningful for $x \ge 3$ — otherwise the hypotenuse is shorter than the given leg, which isn't a valid right triangle.

## Exercise 7

> An automobile travels at 30 miles per hour. Write a formula which expresses the distance $d$ traveled in feet as a function of the time $t$ which represents the number of seconds of travel.

### My answer

First I need the speed in the target units: **feet per second**.

$$30\ \tfrac{\text{mi}}{\text{hr}} \;\cdot\; \frac{5280\ \text{ft}}{1\ \text{mi}} \;\cdot\; \frac{1\ \text{hr}}{3600\ \text{s}} \;=\; \frac{30 \cdot 5280}{3600}\ \tfrac{\text{ft}}{\text{s}} \;=\; 44\ \tfrac{\text{ft}}{\text{s}}.$$

At constant speed, distance = speed × time:

$$\boxed{\,d = 44\, t\,}\quad (d \text{ in feet},\ t \text{ in seconds}).$$

This is linear and single-valued.

## Exercise 8

> A rectangle is required to have an area of 4 square feet but its dimensions may vary. If one side has length $x$, express the perimeter $p$ of the rectangle as a function of $x$.

### My answer

The area constraint pins down the other side. If one side is $x$ and the other side is $y$, then $xy = 4$, so

$$y = \frac{4}{x}.$$

The perimeter of a rectangle is twice the sum of its sides:

$$p = 2(x + y) = 2\!\left(x + \frac{4}{x}\right) = 2x + \frac{8}{x}.$$

So

$$\boxed{\,p(x) = 2x + \tfrac{8}{x}\,}\quad (x > 0).$$

**A few observations that will matter once we start doing derivatives:**

- As $x \to 0^+$, the rectangle becomes a very thin sliver: one side is tiny, the other is huge, and $p \to \infty$.
- As $x \to \infty$, the *other* side is tiny, and again $p \to \infty$.
- Somewhere in the middle $p$ must attain a minimum — the shape that minimizes perimeter for a given area. We can already guess it's the square $x = 2$ (giving $y = 2$ and $p = 8$), and calculus will confirm this by setting $p'(x) = 0$.

## What I'm taking away from Section 2.1

This section wasn't really about the derivative yet — it's Kline making sure we're solid on functions before he defines rate of change in the next section. The recurring theme in these eight exercises:

1. **Identifying independent vs. dependent variable** is a modeling decision, not just an algebra decision. Which quantity are we *choosing*? Which one is *determined*?
2. **Solving for the other variable can break single-valuedness.** Inverting anything with an even power (Exercises 3 and 5) forces us to face the $\pm$.
3. **Physical context restores single-valuedness.** Radii, lengths, and side lengths are nonnegative, so we pick the positive branch and move on.
4. **Constraints turn multi-variable problems into single-variable functions** (Exercise 8): the area constraint $xy = 4$ let me express $p$ purely as a function of $x$. This is exactly the setup calculus will use for optimization.
