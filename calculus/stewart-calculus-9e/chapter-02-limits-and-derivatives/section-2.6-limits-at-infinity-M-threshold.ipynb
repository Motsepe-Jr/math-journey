# Section 2.6: Limits at Infinity From First Principles

## From the $\delta$ Control Knob to the $M$ Threshold Switch

In Section 2.4 we practiced the $\epsilon$-$\delta$ game for limits at a specific point. Now we shift gears: we want to make precise what it means for a function to approach a value as $x$ runs off to infinity.

The worked example throughout is

$$\lim_{x \to \infty} \frac{1}{x-3} = 0.$$

---

## What Changes When $x \to \infty$

The game changes completely here. When $x$ goes to infinity, we are no longer using a tiny control knob ($\delta$) around a specific number. Instead, think of our new control knob as a **giant threshold switch** called $M$.

### The New First-Principles Intuition

Instead of promising to keep $x$ *close* to a point, we are promising that we can force the output as close to $0$ as anyone wants (within $\epsilon$), just by pushing $x$ **far enough to the right** (past some massive number $M$).

Our new goal is to find an if-then rule:

> If $\;x > M,\;$ then $\;\left|\dfrac{1}{x-3} - 0\right| < \epsilon.$

So the roles change from Section 2.4 like this:

| Limit at a point | Limit at infinity |
|---|---|
| $\epsilon$ is the vertical tolerance | $\epsilon$ is *still* the vertical tolerance |
| $\delta$ is horizontal *closeness to a point* | $M$ is horizontal *distance out to the right* |
| $0 < \lvert x - a \rvert < \delta$ | $x > M$ |

Let us break down the strategy to find $M$ from first principles, without assuming anything.

---

## Step-by-Step Breakdown

### Step 1: Analyze the Error Target

We want the distance between our function and $0$ to be less than $\epsilon$:

$$\left|\frac{1}{x-3} - 0\right| < \epsilon \quad\Longrightarrow\quad \left|\frac{1}{x-3}\right| < \epsilon.$$

Since we are letting $x$ push toward positive infinity, we can safely assume $x$ is much larger than $3$. This means $x - 3$ is a positive number, so we can drop the absolute value bars completely:

$$\frac{1}{x-3} < \epsilon.$$

### Step 2: Isolate the Input $x$

We want to figure out what $x$ needs to be bigger than. Solve for $x$:

Take the reciprocal of both sides (remembering that flipping fractions flips the inequality sign, since both sides are positive):

$$x - 3 > \frac{1}{\epsilon}.$$

Add $3$ to both sides:

$$x > \frac{1}{\epsilon} + 3.$$

### Step 3: Setting the Threshold Switch ($M$)

Look at that beautiful result. It tells you exactly how far right you need to push $x$. If you set your giant threshold switch to

$$M = \frac{1}{\epsilon} + 3,$$

then as long as $x > M$, your error is guaranteed to be smaller than $\epsilon$.

No "worst-case scenario" bounding or safety nets are needed here — the fact that $x$ is getting infinitely large already does that work for us.

---

## The Formal Proof

Let us write it out formally to see it work in reverse.

**Proof.** Given any $\epsilon > 0$, let $M = \dfrac{1}{\epsilon} + 3$. Assume $x > M$.

Then

$$x > \frac{1}{\epsilon} + 3 \quad\Longrightarrow\quad x - 3 > \frac{1}{\epsilon}.$$

Since $\epsilon > 0$, the quantity $\dfrac{1}{\epsilon}$ is positive, which forces $x - 3 > 0$. So we may safely take the reciprocal of both sides and flip the inequality sign:

$$\frac{1}{x-3} < \epsilon.$$

Because $x - 3 > 0$, we have $\dfrac{1}{x-3} = \left|\dfrac{1}{x-3}\right|$. Therefore

$$\left|\frac{1}{x-3} - 0\right| = \left|\frac{1}{x-3}\right| < \epsilon.$$

This completes the proof. $\blacksquare$

---

## See How It Works With Real Numbers

If someone sets a tight error tolerance of $\epsilon = 0.01$, you flip the switch:

$$M = \frac{1}{0.01} + 3 = 100 + 3 = 103.$$

As long as they pick an $x > 103$ (say, $x = 1003$), the output is

$$\frac{1}{1003 - 3} = \frac{1}{1000} = 0.001,$$

which is well within their $\epsilon = 0.01$ error target.

Switching the control knob from a tiny interval ($\delta$) to a giant wall ($M$) is really the whole idea of a limit at infinity: instead of *pinning* $x$ near a point, we *cross a horizontal line* and stay past it.

```python
# Setup
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = [10, 6]
plt.rcParams['font.size'] = 12
```

```python
# Verify the rule M = 1/eps + 3 numerically for a few tolerances.

def f(x):
    return 1.0 / (x - 3)

def M_of(eps):
    return 1.0 / eps + 3.0

eps_vals = [0.1, 0.01, 0.001, 1e-6]

rows = []
for eps in eps_vals:
    M = M_of(eps)
    # Sample a point just past M and one well past M
    x_edge = M + 1e-6                     # essentially at the boundary
    x_far  = M * 10                        # 10x past the threshold
    rows.append({
        'epsilon': eps,
        'M = 1/eps + 3': M,
        'x_edge': x_edge,
        '|f(x_edge)|': abs(f(x_edge)),
        'edge < eps?': abs(f(x_edge)) < eps,
        'x_far': x_far,
        '|f(x_far)|': abs(f(x_far)),
        'far < eps?': abs(f(x_far)) < eps,
    })

df = pd.DataFrame(rows)
with pd.option_context('display.float_format', lambda v: f'{v:.6g}'):
    print(df.to_string(index=False))
```

```python
# Visualize: pick eps = 0.1, draw f(x), the +/- eps band, and the M cutoff.

eps = 0.1
M = M_of(eps)

x = np.linspace(4, 40, 800)
y = f(x)

fig, ax = plt.subplots()
ax.plot(x, y, color='steelblue', label=r'$f(x) = \dfrac{1}{x-3}$')

# Horizontal asymptote y = 0
ax.axhline(0, color='black', lw=0.8)

# The epsilon band around 0
ax.axhspan(-eps, eps, color='orange', alpha=0.15,
           label=fr'$\epsilon$-band  $|y| < {eps}$')
ax.axhline( eps, color='orange', lw=0.8, ls='--')
ax.axhline(-eps, color='orange', lw=0.8, ls='--')

# The threshold M
ax.axvline(M, color='crimson', lw=1.2, ls='--',
           label=fr'$M = 1/\epsilon + 3 = {M:g}$')
ax.axvspan(M, x[-1], color='crimson', alpha=0.05)

ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.set_title(r'Past $x = M$, the curve is trapped inside the $\epsilon$-band')
ax.legend(loc='upper right')
ax.set_ylim(-0.05, 0.7)
plt.show()

print(f'At the boundary x = M = {M}:  f(M) = {f(M):.6f}  (compare to eps = {eps})')
```

## Practice Quiz: Finding $M$

Same framework, different functions. For each problem, do the same three moves:

1. Write the error inequality $|f(x) - L| < \epsilon$ and (for $x$ large enough) drop the absolute value.
2. Isolate $x$ to get $x > (\text{something in } \epsilon)$.
3. Read off $M$ from the right-hand side.

Work each one out on paper first, then scroll to the solutions below to check.

**Problem 1.** Prove $\displaystyle \lim_{x \to \infty} \frac{1}{x} = 0$. Find a formula for $M$ in terms of $\epsilon$.

**Problem 2.** Prove $\displaystyle \lim_{x \to \infty} \frac{1}{x+5} = 0$. Find a formula for $M$ in terms of $\epsilon$.

**Problem 3.** Prove $\displaystyle \lim_{x \to \infty} \frac{2}{x-1} = 0$. Find a formula for $M$ in terms of $\epsilon$.

**Problem 4.** Prove $\displaystyle \lim_{x \to \infty} \frac{1}{\sqrt{x}} = 0$. Find a formula for $M$ in terms of $\epsilon$. (This one is slightly different — the algebra involves squaring.)

**Problem 5 (stretch).** Prove $\displaystyle \lim_{x \to \infty} \frac{3x + 1}{x} = 3$. Find a formula for $M$ in terms of $\epsilon$. (Simplify $\left|\frac{3x+1}{x} - 3\right|$ first.)

### Solutions

**Problem 1.**  $\left|\dfrac{1}{x} - 0\right| = \dfrac{1}{x} < \epsilon$ (for $x > 0$) gives $x > \dfrac{1}{\epsilon}$. So $M = \dfrac{1}{\epsilon}$.

**Problem 2.**  $\dfrac{1}{x+5} < \epsilon$ (both sides positive for $x$ large) gives $x + 5 > \dfrac{1}{\epsilon}$, so $x > \dfrac{1}{\epsilon} - 5$. Take $M = \dfrac{1}{\epsilon} - 5$ (or, cleanly, $M = \dfrac{1}{\epsilon}$ — even bigger is fine).

**Problem 3.**  $\dfrac{2}{x-1} < \epsilon$ gives $x - 1 > \dfrac{2}{\epsilon}$, so $x > \dfrac{2}{\epsilon} + 1$. Take $M = \dfrac{2}{\epsilon} + 1$.

**Problem 4.**  $\dfrac{1}{\sqrt{x}} < \epsilon$ gives $\sqrt{x} > \dfrac{1}{\epsilon}$, so $x > \dfrac{1}{\epsilon^{2}}$. Take $M = \dfrac{1}{\epsilon^{2}}$. Notice: to shrink the error by a factor of $10$, you have to push $x$ out by a factor of $100$ — square-root decay is *slower* than $1/x$ decay.

**Problem 5.**  Simplify first: $\left|\dfrac{3x+1}{x} - 3\right| = \left|\dfrac{3x + 1 - 3x}{x}\right| = \dfrac{1}{x}$ (for $x > 0$). Same as Problem 1: $M = \dfrac{1}{\epsilon}$.

```python
# Auto-check the practice solutions numerically.
# For each problem: pick an eps, compute the claimed M, then verify that
# the error is < eps at x = M + small and comfortably < eps far past M.

problems = [
    ('P1: 1/x -> 0',           lambda x: 1/x,           0,   lambda e: 1/e),
    ('P2: 1/(x+5) -> 0',       lambda x: 1/(x+5),       0,   lambda e: 1/e - 5),
    ('P3: 2/(x-1) -> 0',       lambda x: 2/(x-1),       0,   lambda e: 2/e + 1),
    ('P4: 1/sqrt(x) -> 0',     lambda x: 1/np.sqrt(x),  0,   lambda e: 1/e**2),
    ('P5: (3x+1)/x -> 3',      lambda x: (3*x + 1)/x,   3,   lambda e: 1/e),
]

eps = 0.01

rows = []
for name, f_prob, L, M_rule in problems:
    M = M_rule(eps)
    # Guard against a negative M (e.g. P2 with large eps): boundary test only makes
    # sense for x > 0, so clamp for the sample point.
    x_edge = max(M, 1) + 1e-3
    x_far  = max(M, 1) * 10
    rows.append({
        'problem': name,
        'eps': eps,
        'M': M,
        '|f(M+)-L|': abs(f_prob(x_edge) - L),
        'edge < eps?': abs(f_prob(x_edge) - L) < eps,
        '|f(10M)-L|': abs(f_prob(x_far) - L),
        'far < eps?': abs(f_prob(x_far) - L) < eps,
    })

with pd.option_context('display.float_format', lambda v: f'{v:.6g}'):
    print(pd.DataFrame(rows).to_string(index=False))
```

# Third Variant: An Infinite Limit at a Specific Point

$$\lim_{x \to 3^+} \frac{1}{x-3} = \infty$$

Now we are dealing with an **infinite limit at a specific point**. Because the output explodes to infinity, there is no target error ($\epsilon$) on the $y$-axis anymore. Instead, the game changes to a **high-score challenge**.

## The Infinite Limit Intuition

Think of this as a game where a challenger picks a massive, skyscraper-high value on the vertical $y$-axis and names it $M$ (the target high score). They challenge you:

> *"Can you force the output of your function to blow past my high score $M$?"*

You say:

> *"Absolutely. I can make the output beat your score $M$, as long as I twist my horizontal control knob $\delta$ so that $x$ crawls incredibly close to $3$ from the right side ($3^+$)."*

Mathematically, we want to find a rule:

> If $\;3 < x < 3 + \delta,\;$ then $\;\dfrac{1}{x-3} > M.$

*(Note: we write $3 < x < 3 + \delta$ instead of using absolute values because $3^+$ means $x$ is strictly sneaking in from the right side, so $x$ is always slightly bigger than $3$.)*

---

## Step-by-Step Breakdown

### Step 1: Analyze the High-Score Goal

We want our output to beat the challenger's giant target score $M$:

$$\frac{1}{x-3} > M.$$

We assume $M > 0$ because they are challenging us with a large positive number.

### Step 2: Isolate the Control Knob Variable ($x - 3$)

We need to see how close $x$ has to be to $3$. Flip both sides of the inequality to get $x - 3$ out of the denominator. Remember: flipping positive fractions flips the inequality sign.

$$x - 3 < \frac{1}{M}.$$

### Step 3: Setting the Control Knob ($\delta$)

Look at that elegant result. It tells us exactly how close $x$ needs to get to $3$. The distance between $x$ and $3$ is just $x - 3$. If we set our control knob to

$$\delta = \frac{1}{M},$$

then as long as $x$ is within $\delta$ units to the right of $3$, the denominator becomes so small that the fraction is guaranteed to explode past the challenger's high score $M$.

---

## The Formal Proof

Let us reverse the reasoning to write out the airtight argument.

**Proof.** Given any $M > 0$, let $\delta = \dfrac{1}{M}$. Suppose $3 < x < 3 + \delta$.

Subtract $3$ from all parts of the inequality to isolate the distance from $3$:

$$0 < x - 3 < \delta.$$

Substitute $\delta = \dfrac{1}{M}$:

$$0 < x - 3 < \frac{1}{M}.$$

Since both $x - 3$ and $\dfrac{1}{M}$ are strictly positive, taking the reciprocal of both sides flips the inequality:

$$\frac{1}{x-3} > M.$$

Therefore $\dfrac{1}{x-3} > M$, completing the proof. $\blacksquare$

```python
# Numerically verify the rule delta = 1/M across several challenger high-scores.

def f_inf(x):
    return 1.0 / (x - 3)

M_vals = [10, 100, 1_000, 1_000_000]

rows = []
for M in M_vals:
    delta = 1.0 / M
    # Pick a point strictly inside (3, 3 + delta) and at the very edge
    x_inside = 3 + delta / 2
    x_edge   = 3 + delta            # boundary: f(x) should equal M exactly
    rows.append({
        'M (challenger)': M,
        'delta = 1/M': delta,
        'x_inside = 3 + delta/2': x_inside,
        'f(x_inside)': f_inf(x_inside),
        'inside beats M?': f_inf(x_inside) > M,
        'x_edge = 3 + delta': x_edge,
        'f(x_edge)': f_inf(x_edge),
    })

with pd.option_context('display.float_format', lambda v: f'{v:.6g}'):
    print(pd.DataFrame(rows).to_string(index=False))
```

```python
# Visualize: pick M = 20 (high-score), find delta = 1/M, and show that inside
# the vertical strip 3 < x < 3 + delta the curve is above the horizontal line y = M.

M_hi = 20.0
delta_hi = 1.0 / M_hi   # = 0.05

x = np.linspace(3.001, 3.5, 800)
y = 1.0 / (x - 3)

fig, ax = plt.subplots()
ax.plot(x, y, color='steelblue', label=r'$f(x) = \dfrac{1}{x-3}$')

# The vertical asymptote at x = 3
ax.axvline(3, color='black', lw=0.8, ls=':')

# The challenger's high-score line y = M
ax.axhline(M_hi, color='crimson', lw=1.2, ls='--',
           label=fr'challenger: $y = M = {M_hi:g}$')
ax.axhspan(M_hi, 100, color='crimson', alpha=0.05)

# The response: delta strip on the right of x = 3
ax.axvspan(3, 3 + delta_hi, color='orange', alpha=0.20,
           label=fr'response: $3 < x < 3+\delta$,  $\delta = 1/M = {delta_hi:g}$')

ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.set_title(r'Inside the $\delta$-strip on the right of $3$, the curve beats the $M$ line')
ax.set_xlim(2.98, 3.30)
ax.set_ylim(0, 80)
ax.legend(loc='upper right')
plt.show()

# Sanity check: at the boundary x = 3 + delta, f(x) should equal exactly M.
x_edge = 3 + delta_hi
print(f'At x = 3 + delta = {x_edge}:   f(x) = {1/(x_edge - 3):.4f}   (target M = {M_hi})')
print(f'A bit closer, x = 3 + delta/2:  f(x) = {1/(delta_hi/2):.4f}   (should exceed M)')
```

## The Three Limit Games — A Unified Summary

We have now walked through all three fundamental configurations of limit proofs from first principles.

| Limit type | The challenger sets... | You control... | Core math strategy |
|---|---|---|---|
| **At a point:**  $\displaystyle \lim_{x \to c} f(x) = L$ | Tiny target error ($\epsilon$) on $y$ | Tiny input window ($\delta$) around $c$ | Keep $\lvert x - c \rvert < \text{something with } \epsilon$ |
| **At infinity:**  $\displaystyle \lim_{x \to \infty} f(x) = L$ | Tiny target error ($\epsilon$) on $y$ | Giant right-hand wall ($M$) on $x$ | Push $x > \text{something with } \epsilon$ |
| **To infinity:**  $\displaystyle \lim_{x \to c} f(x) = \infty$ | Giant high score ($M$) on $y$ | Tiny input window ($\delta$) around $c$ | Keep $\lvert x - c \rvert < \text{something with } \tfrac{1}{M}$ |

**The single underlying pattern:**

1. Write down the goal the challenger imposes (either $|f(x) - L| < \epsilon$ or $f(x) > M$).
2. Do algebra until the input variable is isolated.
3. Read the required control ($\delta$ or $M$) directly off the right-hand side.
4. If the algebra leaves a wild expression in $x$, install a safety net (Section 2.4, Step 2) first, then continue.

That is the whole game. Everything after this in calculus — derivatives, integrals, continuity — sits on top of this single mental move: *given a challenge on the output, produce a control on the input that guarantees it.*
