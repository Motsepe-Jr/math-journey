This chapter is about counting the number of ways to arrange or select things. The main objects of study are **lists** (ordered sequences of symbols).

A list of length $k$ drawn from a set of $n$ symbols looks like:

$$(\_, \_, \_, \ldots, \_) \quad \leftarrow k \text{ slots}$$

Two key distinctions:
- **Repetition allowed** — the same symbol can appear more than once
- **Repetition not allowed** (non-repetitive list) — each symbol can appear at most once

---
## Lists with Repetition Allowed

**Rule:** To form a list of length $k$ from a set of $n$ symbols where repetition is allowed, there are $n$ choices for each slot:

$$\underbrace{n \cdot n \cdot n \cdots n}_{k \text{ times}} = n^k$$

**Example 3.2(a):** Lists of length 4 from $\{A, B, C, D, E, F, G\}$ (7 symbols), repetition allowed:

$$7 \cdot 7 \cdot 7 \cdot 7 = 7^4 = 2401$$

```python
n = 7
k = 4
total_with_rep = n**k
print(f"Lists of length {k} from {n} symbols (repetition allowed): {n}^{k} = {total_with_rep}")
```

---
## Lists Without Repetition (Non-Repetitive Lists)

**Rule:** To form a list of length $k$ from $n$ symbols where repetition is **not** allowed, each slot has one fewer choice than the previous:

$$n \cdot (n-1) \cdot (n-2) \cdots (n-k+1)$$

This is because once a symbol is placed in a slot, it can no longer be used in later slots.

**Example 3.2(b):** Lists of length 4 from 7 symbols, **no repetition**:

$$7 \cdot 6 \cdot 5 \cdot 4 = 840$$

Visualizing the slots:

| Slot 1 | Slot 2 | Slot 3 | Slot 4 |
|--------|--------|--------|--------|
| 7 choices | 6 choices | 5 choices | 4 choices |

```python
from math import prod

n, k = 7, 4

# No repetition: n * (n-1) * (n-2) * ... * (n-k+1)
no_rep = prod(range(n, n - k, -1))
print(f"Lists of length {k} from {n} symbols (no repetition): {' * '.join(str(x) for x in range(n, n-k, -1))} = {no_rep}")
```

## Complement Counting (Divide and Conquer)

> **Key idea:** Sometimes it is easier to count everything **except** what you want, and then subtract.
>
> $$\text{count}(\text{desired}) = \text{count}(\text{total}) - \text{count}(\text{not desired})$$

This is especially useful when the "not desired" case is simpler to count than the "desired" case directly.

---

### Example 3.2(c): No repetition, list must contain E

**Direct approach (by cases):** E must appear in exactly one of the 4 slots. Divide into 4 types based on where E lands:

| Type | Slot 1 | Slot 2 | Slot 3 | Slot 4 |
|------|--------|--------|--------|--------|
| Type 1 | **E** | 6 choices | 5 choices | 4 choices |
| Type 2 | 6 choices | **E** | 5 choices | 4 choices |
| Type 3 | 6 choices | 5 choices | **E** | 4 choices |
| Type 4 | 6 choices | 5 choices | 4 choices | **E** |

Each type gives $6 \cdot 5 \cdot 4 = 120$ lists. Total:

$$4 \times 120 = 480$$

---

### Example 3.2(d): Repetition allowed, list must contain E

**Complement approach:** Counting directly is hard because E could appear 1, 2, 3, or 4 times. Instead:

$$\text{lists with at least one E} = \underbrace{7^4}_{\text{total}} - \underbrace{6^4}_{\text{lists with NO E}}$$

$$= 2401 - 1296 = 1105$$

The "$6^4$ lists with no E" are built by choosing freely from the 6 symbols $\{A,B,C,D,F,G\}$ — E is simply excluded from all choices.

> **Why is complement easier here?** Because "at least one E" includes many sub-cases (E once, E twice, E three times, E four times), but "no E at all" is just one clean case.

```python
n = 7       # total symbols
k = 4       # list length
# No repetition, must contain E — direct by cases
e_no_rep = 4 * prod(range(n - 1, n - 1 - (k - 1), -1))   # 4 * 6*5*4
print(f"(c) No repetition, must contain E: 4 × {6}×{5}×{4} = {e_no_rep}")

# Repetition allowed, must contain E — complement
total = n**k
no_e  = (n - 1)**k
e_with_rep = total - no_e
print(f"\n(d) Repetition allowed, must contain E:")
print(f"    total={total}, no-E={no_e}, answer={e_with_rep}")
```

## Factorials

The **factorial** of a non-negative integer $n$, written $n!$, is the product of all positive integers up to $n$:

$$n! = n \cdot (n-1) \cdot (n-2) \cdots 2 \cdot 1$$

Special case: $0! = 1$ (by convention, and it makes formulas work out cleanly).

| $n$ | $n!$ |
|-----|------|
| 0 | 1 |
| 1 | 1 |
| 2 | 2 |
| 3 | 6 |
| 4 | 24 |
| 5 | 120 |
| 6 | 720 |
| 7 | 5040 |

**Connection to lists:** The number of non-repetitive lists of length $k$ from $n$ symbols can be written neatly using factorials:

$$n \cdot (n-1) \cdots (n-k+1) = \frac{n!}{(n-k)!}$$

For $k = n$ (arrange **all** symbols), this becomes $n!$ — the number of ways to arrange $n$ distinct objects in a row.

```python
from math import factorial

# Print factorial table
print(f"{'n':>4}  {'n!':>10}")
print("-" * 18)
for n in range(11):
    print(f"{n:>4}  {factorial(n):>10}")

# Verify: n!/(n-k)! = n*(n-1)*...*(n-k+1)
n, k = 7, 4
via_factorial = factorial(n) // factorial(n - k)
via_product   = prod(range(n, n - k, -1))
print(f"\n7! / (7-4)! = {factorial(7)} / {factorial(3)} = {via_factorial}")
print(f"7×6×5×4    = {via_product}")
print(f"Equal? {via_factorial == via_product}")
```

# Section 3.4 — Factorials and Permutations

The previous cells already introduced $n!$ as a formula. §3.4 tightens the story: it defines the factorial *combinatorially*, defines what a **permutation** actually is, and then generalizes to **$k$-permutations** with the notation $P(n, k)$.

## Definition 3.1 (Factorial, combinatorial version)

> If $n$ is a non-negative integer, then $n!$ is the number of lists of length $n$ that can be made from $n$ symbols, **without repetition**.
>
> Thus $0! = 1$ and $1! = 1$. If $n > 1$, then $n! = n(n-1)(n-2) \cdots 3 \cdot 2 \cdot 1$.

So $n!$ is not just a formula — it *counts something concrete*. It's the number of ways to line up $n$ distinct symbols in a row with no repeats.

### Why $0! = 1$ is not a trick

Students often want to say $0! = 0$. It is not. There is exactly **one** length-$0$ list from $0$ symbols: the empty list $()$. So $0! = 1$ by definition, and every table in the book agrees.

A second confirmation comes from the recursion

$$n! \;=\; n \cdot (n-1)!. \tag{3.1}$$

Plugging in $n = 1$ gives $1! = 1 \cdot 0!$, so $1 = 1 \cdot 0!$, forcing $0! = 1$. If you insisted $0! = 0$, you would get $1! = 0$, which is nonsense.

## Permutations of a Set

> A **permutation** of a set is an arrangement of *all* of the set's elements in a row — a non-repetitive list that uses every element.

For $X = \{1, 2, 3\}$ the permutations are

$$123,\ 132,\ 213,\ 231,\ 312,\ 321.$$

There are $3! = 6$ of them, which matches Definition 3.1 exactly.

**A set with $n$ elements has $n!$ permutations.** Some examples:

- $\{1, 2, 3\}$ → $3! = 6$ arrangements
- The four queens $\{Q\clubsuit, Q\diamondsuit, Q\heartsuit, Q\spadesuit\}$ → $4! = 24$
- Letters $\{a, b, c, d, e, f, g\}$ → $7! = 5040$
- $20$ students to $20$ desks → $20! \approx 2.43 \times 10^{18}$

*Aside:* "arrangement in a row" is a slight abuse. A classroom with 4 rows of 5 desks still counts as a permutation of the 20 students because we can *number* the desks $1$ through $20$ and read off the list.

```python
from itertools import permutations
from math import factorial

X = ('1', '2', '3')

# List all permutations of X
perms = list(permutations(X))
print(f"All permutations of {{1, 2, 3}} — there are {len(perms)} = 3! = {factorial(3)}:")
print("  ", ",  ".join("".join(p) for p in perms))

# Recursion check: n! = n · (n-1)!
print("\nSanity check of  n! = n · (n-1)!")
for m in range(1, 8):
    lhs = factorial(m)
    rhs = m * factorial(m - 1)
    print(f"  {m}! = {lhs:>5}    vs.   {m} · {m-1}! = {m} · {factorial(m-1)} = {rhs:>5}   ok? {lhs == rhs}")
```

## Example 3.8 — Lists of Length Seven from $\{a, b, c, d, e, f, g\}$

**(a) How many such lists are there if repetition is not allowed?**

Seven slots, seven distinct letters, no repeats:

$$7! = 5040.$$

**(b) How many such lists are there if repetition is not allowed and the first two entries must be vowels?**

The set contains **two** vowels ($a, e$) and **five** consonants ($b, c, d, f, g$). Slots 1 and 2 must hold the vowels (in some order), and slots 3–7 must hold the consonants (in some order). By the multiplication principle:

$$\underbrace{2 \cdot 1}_{\text{vowels}} \cdot \underbrace{5 \cdot 4 \cdot 3 \cdot 2 \cdot 1}_{\text{consonants}} = 2! \cdot 5! = 2 \cdot 120 = 240.$$

**(c) How many such lists are there in which repetition IS allowed, and the list must contain at least one repeated letter?**

This is a natural place for the **subtraction principle**:

- Let $U$ = all length-$7$ lists from $7$ letters with repetition allowed. $|U| = 7^7 = 823{,}543$.
- Let $X \subseteq U$ = those lists with **no** repetition. $|X| = 7! = 5040$.
- Then lists with **at least one repeat** are $U \setminus X$, and

$$|U \setminus X| = |U| - |X| = 7^7 - 7! = 823{,}543 - 5{,}040 = 818{,}503.$$

**Takeaway:** "At least one repeat" scatters into many messy sub-cases; "no repeats" is one clean case. Subtract instead of counting directly.

```python
from math import factorial

letters = ('a', 'b', 'c', 'd', 'e', 'f', 'g')
vowels     = {'a', 'e'}
consonants = set(letters) - vowels
n = len(letters)   # 7

# (a) length-7 lists, no repetition
part_a = factorial(n)
print(f"(a) 7!                            = {part_a}")

# (b) no repetition; first two entries must be vowels
#     vowel choices: 2 * 1 for the first two slots
#     consonant choices: 5 * 4 * 3 * 2 * 1 for the last five slots
part_b = factorial(len(vowels)) * factorial(len(consonants))    # 2! * 5!
print(f"(b) 2! · 5!                       = {factorial(2)} · {factorial(5)} = {part_b}")

# (c) repetition allowed, MUST contain at least one repeat
#     complement: total minus lists with no repetition
total = n ** n
no_rep = factorial(n)
part_c = total - no_rep
print(f"(c) 7^7 − 7!                      = {total:,} − {no_rep} = {part_c:,}")
```

## $k$-Permutations and $P(n, k)$

A **$k$-permutation** of $X$ is a *non-repetitive list of length $k$* made from elements of $X$. In plain terms: take $k$ elements from $X$ (no repeats) and arrange them in a row.

Notice a permutation of $X$ (the previous idea) is just the special case $k = |X|$ — arrange **all** of them.

For $X = \{a, b, c, d\}$ ($n = 4$):

| $k$ | What are the $k$-permutations? | Count |
|---|---|---|
| $0$ | The empty list $()$ only | $1$ |
| $1$ | $a,\ b,\ c,\ d$ | $4$ |
| $2$ | $ab, ac, ad, ba, bc, bd, ca, cb, cd, da, db, dc$ | $12 = 4 \cdot 3$ |
| $3$ | $abc, acb, bac, bca, cab, cba,\ \ldots$ (24 total) | $24 = 4 \cdot 3 \cdot 2$ |
| $4$ | All permutations of $X$ | $24 = 4!$ |
| $5$ | *None* — can't make a length-$5$ non-repetitive list from $4$ symbols | $0$ |

### The notation $P(n, k)$

$P(n, k)$ denotes the number of $k$-permutations of an $n$-element set. From the table: $P(4, 0) = 1$, $P(4, 1) = 4$, $P(4, 2) = 12$, $P(4, 3) = 24$, $P(4, 4) = 24$, $P(4, 5) = 0$.

### The formula (via the multiplication principle)

When we build a non-repetitive length-$k$ list from $n$ symbols, we have $n$ choices for slot 1, $n - 1$ for slot 2, $n - 2$ for slot 3, and so on. The $i$-th slot has $n - i + 1$ choices, so the last (slot $k$) has $n - k + 1$ choices. Multiplying:

$$P(n, k) = n \cdot (n-1) \cdot (n-2) \cdots (n - k + 1). \tag{3.2}$$

That's $k$ factors in the product.

**Examples:** $P(10, 3) = 10 \cdot 9 \cdot 8 = 720$; $\; P(10, 10) = 10! = 3{,}628{,}800$; $\; P(10, 11) = 10 \cdot 9 \cdots 1 \cdot 0 = 0$ (the 11th factor is $0$, which kills the whole product — reassuringly, since you *can't* make an $11$-long non-repetitive list from $10$ symbols).

### The clean factorial form

For $0 \le k \le n$, multiplying by $\dfrac{(n-k)!}{(n-k)!}$ pads the falling product back out to a full factorial:

$$P(n, k) = n(n-1)(n-2) \cdots (n - k + 1) = \frac{n(n-1) \cdots (n-k+1) \cdot (n-k)(n-k-1) \cdots 1}{(n-k)(n-k-1) \cdots 1} = \frac{n!}{(n-k)!}.$$

Quick sanity check: $P(8, 5) = 8 \cdot 7 \cdot 6 \cdot 5 \cdot 4 = 6720$, and $\dfrac{8!}{3!} = \dfrac{40320}{6} = 6720.$ ✓

```python
from itertools import permutations
from math import factorial

X = ('a', 'b', 'c', 'd')
n = len(X)

print(f"X = {X},   |X| = {n}\n")
print(f"{'k':>3}  {'formula':<28}  {'P(n,k)':>10}  {'verified by enumeration':>25}")
print("-" * 75)

for k in range(0, n + 2):        # go one past n to see k > n case
    # Formula: P(n,k) = n! / (n-k)! when 0 <= k <= n, else 0
    if 0 <= k <= n:
        p_formula = factorial(n) // factorial(n - k)
    else:
        p_formula = 0

    # Enumeration: list all length-k non-repetitive lists from X
    listed = list(permutations(X, k))
    p_count = len(listed)

    prod_str = "·".join(str(n - i) for i in range(k)) if k > 0 else "(empty product) = 1"
    print(f"{k:>3}  {prod_str:<28}  {p_formula:>10}  {p_count:>25}")

# Show the 12 explicit 2-permutations from the book
print("\nThe 12 two-permutations of {a,b,c,d}:")
print("  ", "  ".join("".join(p) for p in permutations(X, 2)))

# Show the 24 three-permutations
print(f"\nThe {len(list(permutations(X, 3)))} three-permutations of {{a,b,c,d}} (4 rows of 6):")
threes = ["".join(p) for p in permutations(X, 3)]
for i in range(0, len(threes), 6):
    print("  ", "  ".join(threes[i:i+6]))
```

## Fact 3.4 — Summary

> **Fact 3.4.** A $k$-**permutation** of an $n$-element set is a non-repetitive length-$k$ list made from elements of the set. Informally, a $k$-permutation is an arrangement of $k$ of the set's elements in a row.
>
> The number of $k$-permutations of an $n$-element set is denoted $P(n, k)$, and
>
> $$P(n, k) = n(n-1)(n-2) \cdots (n - k + 1).$$
>
> If $\;0 \le k \le n$, then
>
> $$P(n, k) = \frac{n!}{(n-k)!}.$$

**Two sanity checks that fall right out of the formula:**

| Case | Value | Why |
|---|---|---|
| $P(n, 0)$ | $\dfrac{n!}{(n-0)!} = \dfrac{n!}{n!} = 1$ | Only one length-$0$ list: the empty list. |
| $P(0, 0)$ | $\dfrac{0!}{0!} = 1$ | Same reason — one empty list from zero symbols. |
| $P(n, n)$ | $\dfrac{n!}{0!} = n!$ | Arranging all $n$ elements → just a permutation of the whole set. |
| $P(n, k),\ k > n$ | $0$ | The product runs $n \cdot (n-1) \cdots$ past $0$, killing it. |

Hammack's final remark is worth internalizing: we don't strictly *need* the $P(n, k)$ notation — the multiplication and addition principles already do the job. But $P(n, k)$ is a convenient shorthand and it makes formulas involving combinations cleaner later in the chapter.

## Two Worked Examples (from §3.4)

### Example 3.9 — Marathon Podium

> Ten contestants run a marathon. All finish, and there are no ties. How many different possible rankings are there for first-, second-, and third-place?

A ranking of the top three is exactly a $3$-permutation of the $10$ contestants (order matters: gold, silver, bronze are different). So

$$P(10, 3) = 10 \cdot 9 \cdot 8 = 720.$$

### Example 3.10 — Five-Card Lineups: All Red or All Clubs

> Deal five cards off a standard $52$-card deck and line them up in a row. How many lineups consist of **all red cards, or all clubs**?

Break into cases (addition principle since the two cases are disjoint — a lineup can't be simultaneously "all red" and "all clubs" since clubs are black):

- All red: there are $26$ red cards, so the number of length-$5$ lineups is
$$P(26, 5) = 26 \cdot 25 \cdot 24 \cdot 23 \cdot 22 = 7{,}893{,}600.$$
- All clubs: there are $13$ clubs, so
$$P(13, 5) = 13 \cdot 12 \cdot 11 \cdot 10 \cdot 9 = 154{,}440.$$

**Total:** $\;P(26, 5) + P(13, 5) = 8{,}048{,}040.$

```python
from math import factorial

# Example 3.9 — marathon rankings, P(10, 3)
n_runners = 10
podium = factorial(n_runners) // factorial(n_runners - 3)
print(f"Example 3.9  P(10, 3) = 10·9·8 = {10 * 9 * 8}")
print(f"             via factorials 10!/(10-3)! = {podium}")
print(f"             match? {10 * 9 * 8 == podium}")

# Example 3.10 — 5-card lineups: all red OR all clubs
P_26_5 = factorial(26) // factorial(26 - 5)
P_13_5 = factorial(13) // factorial(13 - 5)
total_lineups = P_26_5 + P_13_5

print(f"\nExample 3.10  P(26, 5) = 26·25·24·23·22 = {26*25*24*23*22:>12,}")
print(f"              P(13, 5) = 13·12·11·10·9  = {13*12*11*10*9:>12,}")
print(f"              P(26,5) + P(13,5)          = {total_lineups:>12,}")
print(f"              via factorials             = {P_26_5 + P_13_5:>12,}")
print(f"              match? {26*25*24*23*22 + 13*12*11*10*9 == total_lineups}")
```
